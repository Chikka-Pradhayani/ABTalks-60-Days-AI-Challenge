```python
import numpy as np
from langchain_openai import ChatOpenAI, OpenAIEmbeddings
from langchain_community.vectorstores import FAISS
```

```python
documents = [
    """
    Nexora Technologies was founded in 2024.
    Its headquarters is located in Hyderabad.
    The company currently has 85 employees.
    """,

    """
    Nexora Technologies launched its AI product called NovaMind
    in March 2026. NovaMind is designed for automated document analysis.
    """,

    """
    Nexora Technologies offers three products:
    NovaMind, DataFlow, and SecureVision.
    """,

    """
    The CEO of Nexora Technologies is Arjun Mehta.
    The CTO is Priya Sharma.
    """,

    """
    Nexora Technologies plans to open its second office in Bengaluru
    during 2027.
    """
]
```

```python
for i, doc in enumerate(documents):
    print(f"Document {i + 1}:")
    print(doc)
    print("-" * 50)
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
print("Embedding model loaded successfully!")
```

```python
from langchain_community.vectorstores import FAISS
```

```python
vectorstore = FAISS.from_texts(
    documents,
    embedding=embeddings
)
print("FAISS vector database created successfully!")
```

```python
query = "Who is the CTO of Nexora Technologies?"
results = vectorstore.similarity_search(query, k=3)
for i, result in enumerate(results, 1):
    print(f"\nResult {i}:")
    print(result.page_content)
```

```python
from transformers import pipeline
generator = pipeline(
    "text-generation",
    model="distilgpt2"
)
print("Local LLM loaded successfully!")
```

```python
response = generator(
    "Artificial intelligence is",
    max_new_tokens=30,
    do_sample=False
)

print(response[0]["generated_text"])
```

```python
def generate_without_rag(query):
    prompt = f"Question: {query}\nAnswer:"
    response = generator(
        prompt,
        max_new_tokens=50,
        do_sample=False
    )
    return response[0]["generated_text"]
```

```python
query = "Who is the CTO of Nexora Technologies?"
answer = generate_without_rag(query)
print(answer)
```

```python
def generate_with_rag(query):
    results = vectorstore.similarity_search(query, k=3)
    context = "\n\n".join(
        [result.page_content for result in results]
    )
    prompt = f"""
Use ONLY the information provided in the CONTEXT to answer the QUESTION.

If the answer is not present in the context, say:
"I don't know based on the provided context."
CONTEXT:
{context}
QUESTION:
{query}
ANSWER:
"""
    response = generator(
        prompt,
        max_new_tokens=80,
        do_sample=False
    )
    return response[0]["generated_text"]
```

```python
query = "Who is the CTO of Nexora Technologies?"
answer = generate_with_rag(query)
print(answer)
```

```python
test_queries = [
    "When was Nexora Technologies founded?",
    "Where is Nexora Technologies headquartered?",
    "What is the name of Nexora's AI product?",
    "Who is the CTO of Nexora Technologies?",
    "Where will Nexora Technologies open its second office?"
]
```

```python
for i, query in enumerate(test_queries, 1):
    print(f"{i}. {query}")
```

```python
comparison_results = []
for query in test_queries:
    without_rag = generate_without_rag(query)
    with_rag = generate_with_rag(query)
    comparison_results.append({
        "Question": query,
        "Without RAG": without_rag,
        "With RAG": with_rag
    })
print("Comparison completed!")
```

```python
for i, result in enumerate(comparison_results, 1):
    print("=" * 80)
    print(f"QUESTION {i}")
    print("=" * 80)
    print("\nQuestion:")
    print(result["Question"])
    print("\nWITHOUT RAG:")
    print(result["Without RAG"])
    print("\nWITH RAG:")
    print(result["With RAG"])
    print()
```

```python
import pandas as pd
comparison_df = pd.DataFrame(comparison_results)
comparison_df
```
