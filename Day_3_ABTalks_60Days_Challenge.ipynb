Artificial Intelligence is changing the way people work. AI systems can understand text, process information, and help people make better decision

```python
text = "Artificial Intelligence is changing the way people work. AI systems can understand text, process information, and help people make better decision"
print(text)
```

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize

tokens = word_tokenize(text)

print(tokens)
```

```python
import string

tokens_no_punctuation = [word for word in tokens if word not in string.punctuation]

print(tokens_no_punctuation)
```

Remove Stop Words

```python
nltk.download('stopwords')

from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

clean_tokens = [
    word for word in tokens_no_punctuation
    if word.lower() not in stop_words
]

print(clean_tokens)
```

Bag of word representation

```python
from collections import Counter
bow = Counter(clean_tokens)
print(bow)
```

```python
import pandas as pd

bow_df = pd.DataFrame(
    list(bow.items()),
    columns=["Word", "Frequency"]
)

print(bow_df)
```
