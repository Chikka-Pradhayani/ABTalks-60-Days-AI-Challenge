```python
documents = [
    {
        "id": "doc_1",
        "title": "Artificial Intelligence",
        "content": """
Artificial Intelligence is a field of computer science that focuses on
creating systems capable of performing tasks that normally require human
intelligence. These tasks include learning, reasoning, problem solving,
understanding language, and recognizing patterns.
"""
    },
    {
        "id": "doc_2",
        "title": "Machine Learning",
        "content": """
Machine Learning is a subset of Artificial Intelligence. It allows
computers to learn patterns from data and make predictions or decisions
without being explicitly programmed for every task. Common approaches
include supervised learning, unsupervised learning, and reinforcement
learning.
"""
    },
    {
        "id": "doc_3",
        "title": "Retrieval Augmented Generation",
        "content": """
Retrieval Augmented Generation, commonly called RAG, combines information
retrieval with language generation. A RAG system first retrieves relevant
information from a knowledge base and then provides that information to
a language model so that it can generate a grounded answer.
"""
    }
]

print("Number of documents:", len(documents))
```

```python
def retrieve_documents(query, top_k=3, title_filter=None):

    # Start with all documents
    documents_to_search = langchain_documents

    # Apply metadata filter if provided
    if title_filter:
        documents_to_search = [
            doc for doc in langchain_documents
            if doc.metadata["title"].lower() == title_filter.lower()
        ]

    # Create a temporary FAISS index using filtered documents
    filtered_vectorstore = FAISS.from_documents(
        documents_to_search,
        embeddings
    )

    # Semantic search
    results = filtered_vectorstore.similarity_search_with_score(
        query,
        k=min(top_k, len(documents_to_search))
    )

    retrieved_docs = []

    for doc, score in results:
        retrieved_docs.append({
            "id": doc.metadata["id"],
            "title": doc.metadata["title"],
            "content": doc.page_content,
            "score": float(score)
        })

    return retrieved_docs
```

```python
def run_rag(query, top_k=3, title_filter=None):

    # 1. Retrieve relevant documents
    retrieved_docs = retrieve_documents(
        query=query,
        top_k=top_k,
        title_filter=title_filter
    )

    # 2. Create context
    context = "\n\n".join(
        f"Source ID: {doc['id']}\n"
        f"Title: {doc['title']}\n"
        f"Content: {doc['content']}"
        for doc in retrieved_docs
    )

    # 3. Create grounded prompt
    prompt = GROUNDING_PROMPT.format(
        context=context,
        query=query
    )

    # 4. Generate answer
    answer = generate_answer(prompt)

    # 5. Convert FAISS distance into similarity
    similarities = [
        1 / (1 + doc["score"])
        for doc in retrieved_docs
    ]

    # 6. Find highest similarity
    highest_similarity = max(similarities) if similarities else 0

    # 7. Collect source IDs
    sources = [
        doc["id"]
        for doc in retrieved_docs
    ]

    # 8. Add source citation
    answer += "\n\nSources: " + ", ".join(sources)

    # 9. Add confidence warning
    if highest_similarity < 0.3:
        answer += (
            "\n\nLow confidence: "
            "The retrieved information may not provide strong "
            "support for this answer."
        )

    # 10. Return complete result
    return {
        "answer": answer,
        "sources": sources,
        "confidence": highest_similarity
    }
```

```python
%%writefile rag_pipeline.py

from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.documents import Document
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
from transformers import pipeline


# 1. Knowledge Base

documents = [
    {
        "id": "doc_1",
        "title": "Artificial Intelligence",
        "content": (
            "Artificial Intelligence is the field of computer science "
            "that focuses on creating systems capable of performing "
            "tasks that normally require human intelligence."
        )
    },
    {
        "id": "doc_2",
        "title": "Machine Learning",
        "content": (
            "Machine Learning is a subset of Artificial Intelligence "
            "that enables computers to learn patterns from data and "
            "make predictions or decisions without being explicitly programmed."
        )
    },
    {
        "id": "doc_3",
        "title": "Retrieval Augmented Generation",
        "content": (
            "Retrieval Augmented Generation, or RAG, combines information "
            "retrieval with language generation. Relevant documents are "
            "retrieved first and then provided to a language model as context."
        )
    }
]


# 2. Chunking

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=300,
    chunk_overlap=50
)

chunks = []

for doc in documents:
    doc_chunks = text_splitter.split_text(doc["content"])

    for i, chunk in enumerate(doc_chunks):
        chunks.append({
            "id": f'{doc["id"]}_chunk_{i}',
            "title": doc["title"],
            "content": chunk
        })


# 3. LangChain Documents

langchain_documents = [
    Document(
        page_content=chunk["content"],
        metadata={
            "id": chunk["id"],
            "title": chunk["title"]
        }
    )
    for chunk in chunks
]


# 4. Embeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)


# 5. FAISS

vectorstore = FAISS.from_documents(
    langchain_documents,
    embeddings
)


# 6. Retrieval

def retrieve_documents(query, top_k=3, title_filter=None):

    documents_to_search = langchain_documents

    if title_filter:
        documents_to_search = [
            doc
            for doc in langchain_documents
            if doc.metadata["title"].lower() == title_filter.lower()
        ]

    if not documents_to_search:
        return []

    filtered_vectorstore = FAISS.from_documents(
        documents_to_search,
        embeddings
    )

    results = filtered_vectorstore.similarity_search_with_score(
        query,
        k=min(top_k, len(documents_to_search))
    )

    retrieved_docs = []

    for doc, score in results:
        retrieved_docs.append({
            "id": doc.metadata["id"],
            "title": doc.metadata["title"],
            "content": doc.page_content,
            "score": float(score)
        })

    return retrieved_docs


# 7. Grounding Prompt

GROUNDING_PROMPT = """
Answer the question using ONLY the retrieved context.

Retrieved Context:
{context}

Question:
{query}

Answer:
"""


# 8. Local LLM

generator = pipeline(
    "text-generation",
    model="distilgpt2"
)


# 9. Generate Answer

def generate_answer(prompt):

    result = generator(
        prompt,
        max_new_tokens=100,
        do_sample=False
    )

    generated_text = result[0]["generated_text"]

    return generated_text.strip()


# 10. Complete RAG Pipeline

def run_rag(query, top_k=3, title_filter=None):

    retrieved_docs = retrieve_documents(
        query=query,
        top_k=top_k,
        title_filter=title_filter
    )

    if not retrieved_docs:
        return {
            "answer": "I don't have enough information in the provided knowledge base.",
            "sources": [],
            "confidence": 0.0
        }

    context = "\n\n".join(
        f"Source ID: {doc['id']}\n"
        f"Title: {doc['title']}\n"
        f"Content: {doc['content']}"
        for doc in retrieved_docs
    )

    prompt = GROUNDING_PROMPT.format(
        context=context,
        query=query
    )

    answer = generate_answer(prompt)

    similarities = [
        1 / (1 + doc["score"])
        for doc in retrieved_docs
    ]

    highest_similarity = max(similarities)

    sources = [
        doc["id"]
        for doc in retrieved_docs
    ]

    answer += "\n\nSources: " + ", ".join(sources)

    if highest_similarity < 0.3:
        answer += (
            "\n\nLow confidence: "
            "The retrieved information may not provide strong "
            "support for this answer."
        )

    return {
        "answer": answer,
        "sources": sources,
        "confidence": highest_similarity
    }


print("RAG pipeline created successfully!")
```

```python
test_queries = [
    # Easy
    {
        "query": "What is artificial intelligence?",
        "difficulty": "Easy"
    },
    {
        "query": "What is machine learning?",
        "difficulty": "Easy"
    },
    {
        "query": "What is RAG?",
        "difficulty": "Easy"
    },
    {
        "query": "What does AI focus on?",
        "difficulty": "Easy"
    },
    {
        "query": "What does machine learning learn from?",
        "difficulty": "Easy"
    },

    # Medium
    {
        "query": "How is machine learning related to artificial intelligence?",
        "difficulty": "Medium"
    },
    {
        "query": "How does RAG combine retrieval and generation?",
        "difficulty": "Medium"
    },
    {
        "query": "What is the purpose of retrieving documents in RAG?",
        "difficulty": "Medium"
    },
    {
        "query": "How does machine learning make predictions?",
        "difficulty": "Medium"
    },
    {
        "query": "What type of tasks can artificial intelligence perform?",
        "difficulty": "Medium"
    },

    # Hard
    {
        "query": "Explain how RAG uses retrieved information to improve language generation.",
        "difficulty": "Hard"
    },
    {
        "query": "Compare artificial intelligence and machine learning based only on the knowledge base.",
        "difficulty": "Hard"
    },
    {
        "query": "Why is retrieval important before generating an answer in a RAG system?",
        "difficulty": "Hard"
    },
    {
        "query": "Explain the relationship between AI, machine learning, and RAG.",
        "difficulty": "Hard"
    },
    {
        "query": "What would happen if the required information was not present in the retrieved context?",
        "difficulty": "Hard"
    }
]

print("Total test queries:", len(test_queries))
```

```python
evaluation_results = []

for item in test_queries:
    retrieved = retrieve_documents(item["query"], top_k=3)

    evaluation_results.append({
        "query": item["query"],
        "difficulty": item["difficulty"],
        "sources": [doc["id"] for doc in retrieved],
        "confidence": max(
            [1 / (1 + doc["score"]) for doc in retrieved],
            default=0.0
        )
    })

print("Evaluation completed.")
print("Total results:", len(evaluation_results))
```

```python
import pandas as pd

evaluation_df = pd.DataFrame(evaluation_results)

evaluation_df[
    ["difficulty", "query", "sources", "confidence"]
]
```

```python
def score_retrieval(row):
    query = row["query"].lower()
    sources = " ".join(row["sources"]).lower()

    # Expected source based on the query
    if "machine learning" in query:
        expected = "doc_2"
    elif "rag" in query or "retriev" in query or "generat" in query:
        expected = "doc_3"
    elif "artificial intelligence" in query or "ai" in query:
        expected = "doc_1"
    else:
        expected = None

    if expected and expected in sources:
        return 1
    return 0


evaluation_df["retrieval_score"] = evaluation_df.apply(
    score_retrieval,
    axis=1
)

retrieval_accuracy = evaluation_df["retrieval_score"].mean()

print("Retrieval Quality Score:", round(retrieval_accuracy * 100, 2), "%")
```
