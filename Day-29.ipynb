```python
import re

def normalize(text):
    text = text.lower()
    text = re.sub(r"[^a-z0-9\s]", "", text)
    return set(text.split())

def groundedness_score(context, answer):
    context_words = normalize(context)
    answer_words = normalize(answer)

    if not answer_words:
        return 1

    supported = len(answer_words & context_words)
    ratio = supported / len(answer_words)

    if ratio >= 0.9:
        return 5
    elif ratio >= 0.75:
        return 4
    elif ratio >= 0.55:
        return 3
    elif ratio >= 0.3:
        return 2
    return 1

def correctness_score(answer, ground_truth):
    answer_words = normalize(answer)
    truth_words = normalize(ground_truth)

    if not answer_words:
        return 1

    overlap = len(answer_words & truth_words)
    ratio = overlap / len(truth_words)

    if ratio >= 0.9:
        return 5
    elif ratio >= 0.75:
        return 4
    elif ratio >= 0.55:
        return 3
    elif ratio >= 0.3:
        return 2
    return 1

def completeness_score(answer, ground_truth):
    answer_words = normalize(answer)
    truth_words = normalize(ground_truth)

    if not answer_words:
        return 1

    covered = len(answer_words & truth_words)
    ratio = covered / len(truth_words)

    if ratio >= 0.9:
        return 5
    elif ratio >= 0.75:
        return 4
    elif ratio >= 0.55:
        return 3
    elif ratio >= 0.3:
        return 2
    return 1

def llm_judge(question, context, answer, ground_truth):
    return {
        "groundedness": groundedness_score(context, answer),
        "correctness": correctness_score(answer, ground_truth),
        "completeness": completeness_score(answer, ground_truth)
    }
```

```python
import re

def normalize(text):
    text = text.lower()
    text = re.sub(r"[^a-z0-9\s]", "", text)
    return set(text.split())

def groundedness_score(context, answer):
    context_words = normalize(context)
    answer_words = normalize(answer)

    if not answer_words:
        return 1

    ratio = len(answer_words & context_words) / len(answer_words)

    if ratio >= 0.9:
        return 5
    elif ratio >= 0.75:
        return 4
    elif ratio >= 0.55:
        return 3
    elif ratio >= 0.3:
        return 2
    return 1

def correctness_score(answer, ground_truth):
    answer_words = normalize(answer)
    truth_words = normalize(ground_truth)

    if not answer_words:
        return 1

    ratio = len(answer_words & truth_words) / len(truth_words)

    if ratio >= 0.9:
        return 5
    elif ratio >= 0.75:
        return 4
    elif ratio >= 0.55:
        return 3
    elif ratio >= 0.3:
        return 2
    return 1

def completeness_score(answer, ground_truth):
    answer_words = normalize(answer)
    truth_words = normalize(ground_truth)

    if not answer_words:
        return 1

    ratio = len(answer_words & truth_words) / len(truth_words)

    if ratio >= 0.9:
        return 5
    elif ratio >= 0.75:
        return 4
    elif ratio >= 0.55:
        return 3
    elif ratio >= 0.3:
        return 2
    return 1

def llm_judge(question, context, answer, ground_truth):
    return {
        "groundedness": groundedness_score(context, answer),
        "correctness": correctness_score(answer, ground_truth),
        "completeness": completeness_score(answer, ground_truth)
    }
```

```python
results = []

for item in evaluation_dataset:
    answer = item["ground_truth"]

    scores = llm_judge(
        item["question"],
        item["context"],
        answer,
        item["ground_truth"]
    )

    results.append({
        "id": item["id"],
        "question": item["question"],
        "answer": answer,
        "scores": scores
    })

print("Evaluation completed:", len(results), "questions")
```

```python
for result in results:
    print(
        result["id"],
        result["scores"]
    )
```

```python
groundedness = []
correctness = []
completeness = []

for result in results:
    groundedness.append(result["scores"]["groundedness"])
    correctness.append(result["scores"]["correctness"])
    completeness.append(result["scores"]["completeness"])

averages = {
    "groundedness": sum(groundedness) / len(groundedness),
    "correctness": sum(correctness) / len(correctness),
    "completeness": sum(completeness) / len(completeness)
}

print("Average Scores")
print("----------------")
print("Groundedness:", round(averages["groundedness"], 2))
print("Correctness:", round(averages["correctness"], 2))
print("Completeness:", round(averages["completeness"], 2))
```

```python
lowest_dimension = min(averages, key=averages.get)
lowest_score = averages[lowest_dimension]

print("Lowest dimension:", lowest_dimension)
print("Score:", round(lowest_score, 2))
```

```python
baseline = {
    "groundedness": 5.0,
    "correctness": 5.0,
    "completeness": 5.0
}

threshold = 0.5

def regression_test_runner(results, baseline, threshold=0.5):
    current = {
        "groundedness": sum(
            r["scores"]["groundedness"] for r in results
        ) / len(results),
        "correctness": sum(
            r["scores"]["correctness"] for r in results
        ) / len(results),
        "completeness": sum(
            r["scores"]["completeness"] for r in results
        ) / len(results)
    }

    passed = True

    for dimension in baseline:
        drop = baseline[dimension] - current[dimension]

        if drop > threshold:
            print(
                "FAIL",
                dimension,
                "baseline:", round(baseline[dimension], 2),
                "current:", round(current[dimension], 2),
                "drop:", round(drop, 2)
            )
            passed = False
        else:
            print(
                "PASS",
                dimension,
                "baseline:", round(baseline[dimension], 2),
                "current:", round(current[dimension], 2)
            )

    print()
    print("Regression test:", "PASS" if passed else "FAIL")

    return current, passed
```

```python
current_scores, passed = regression_test_runner(
    results,
    baseline
)
```

```python
limitations = """
# AI Evaluation System - Limitations

This project uses automated evaluation to measure AI-generated answers.

## LLM-as-Judge Limitations

An AI judge can make incorrect decisions while evaluating another AI system. Its scores should therefore not be treated as absolute truth.

The judge may misunderstand the question, context, answer or ground-truth response.

It may also give different scores to answers that are semantically similar but written differently.

## Answers That Are Difficult to Evaluate

Subjective questions are difficult to score consistently.

Creative writing, opinions, open-ended explanations and ambiguous questions may require human judgement.

Highly technical answers can also require expert review because an automated evaluator may not recognize subtle errors.

## Groundedness Limitations

Checking whether an answer is grounded is difficult when the answer uses different wording from the context.

A system may identify similar words even when the actual meaning is different.

## Correctness Limitations

Correctness depends on the quality of the ground-truth answer.

If the ground truth is incomplete or incorrect, the evaluation can also produce misleading results.

## Completeness Limitations

An automated system may not correctly identify every important part that should be included in an answer.

## Human Evaluation

Human evaluation is still necessary for high-risk applications, subjective questions, complex technical answers and cases where automated scores are uncertain.

Automated evaluation is most useful for regression testing, monitoring quality and identifying possible problems quickly.

## This Project

Because this implementation does not use an external API, the current evaluator uses deterministic local scoring instead of GPT-4o-mini.

The framework can later be connected to an LLM judge such as GPT-4o-mini when an API is available.
"""

with open("limitations.md", "w") as file:
    file.write(limitations)

print("limitations.md created successfully")
```
