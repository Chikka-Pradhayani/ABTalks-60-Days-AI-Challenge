```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer("all-MiniLM-L6-v2")
print("Embedding model loaded successfully!")
```

```python
embeddings = model.encode(corpus)
print("Embedding shape:", embeddings.shape)
```

```python
embeddings = np.asarray(embeddings, dtype=np.float32)
print("Data type:", embeddings.dtype)
print("Shape:", embeddings.shape)
```

```python
dimension = embeddings.shape[1]
index = faiss.IndexFlatL2(dimension)
print("Vector dimension:", dimension)
print("FAISS index created!")
```

```python
index.add(embeddings)
print("FAISS index size:", index.ntotal)
```

```python
def semantic_search(query, top_k=3):
    query_embedding = model.encode([query])
    query_embedding = np.asarray(query_embedding, dtype=np.float32)
    distances, indices = index.search(query_embedding, top_k)
    results = []
    for distance, idx in zip(distances[0], indices[0]):
        results.append({
            "document": corpus[idx],
            "distance": float(distance)
        })
    return results
```

```python
results = semantic_search("How do computers learn from data?", top_k=3)
for i, result in enumerate(results, 1):
    print(f"{i}. Distance: {result['distance']:.4f}")
    print(f"   {result['document']}")
    print()
```

```python
import re
def keyword_search(query, corpus, top_k=3):
    query_words = set(re.findall(r"\b\w+\b", query.lower()))
    scores = []
    for idx, document in enumerate(corpus):
        document_words = set(re.findall(r"\b\w+\b", document.lower()))
        score = len(query_words.intersection(document_words))
        scores.append((idx, score))

    # Sort from highest to lowest
    scores.sort(key=lambda x: x[1], reverse=True)
    results = []
    for idx, score in scores[:top_k]:
        results.append({
            "document": corpus[idx],
            "score": score
        })
    return results
```

```python
results = keyword_search(
    "How do computers learn from data?",
    corpus,
    top_k=3
)
for i, result in enumerate(results, 1):
    print(f"{i}. Score: {result['score']}")
    print(f"   {result['document']}")
    print()
```

```python
test_queries = [
    "How do computers learn patterns from examples?",
    "How can I store information in tables?",
    "How do websites communicate with backend systems?",
    "How can applications use computers over the internet?",
    "How can I protect a computer network from attacks?",
    "How do programs handle unexpected errors?",
    "How can a computer understand images?",
    "How do computers manage running programs?",
    "How can I quickly find a specific record in a database?",
    "How can software automatically increase resources when demand grows?"
]
print("Number of test queries:", len(test_queries))
```

```python
comparison_results = []
for query in test_queries:
    semantic_results = semantic_search(query, top_k=3)
    keyword_results = keyword_search(query, corpus, top_k=3)
    comparison_results.append({
        "Query": query,
        "Keyword Result": keyword_results[0]["document"],
        "Keyword Score": keyword_results[0]["score"],
        "Semantic Result": semantic_results[0]["document"],
        "Semantic Distance": round(semantic_results[0]["distance"], 4)
    })
comparison_df = pd.DataFrame(comparison_results)
comparison_df
```

```python
for query in test_queries:
    print("=" * 100)
    print("QUERY:", query)
    print("\nKEYWORD SEARCH")
    keyword_results = keyword_search(query, corpus, top_k=3)
    for i, result in enumerate(keyword_results, 1):
        print(f"{i}. Score: {result['score']} | {result['document']}")
    print("\nSEMANTIC SEARCH")
    semantic_results = semantic_search(query, top_k=3)
    for i, result in enumerate(semantic_results, 1):
        print(f"{i}. Distance: {result['distance']:.4f} | {result['document']}")
    print()
```

```python
analysis_results = []
for query in test_queries:
    keyword_result = keyword_search(query, corpus, top_k=1)[0]
    semantic_result = semantic_search(query, top_k=1)[0]
    analysis_results.append({
        "Query": query,
        "Keyword Score": keyword_result["score"],
        "Keyword Top Result": keyword_result["document"],
        "Semantic Distance": round(semantic_result["distance"], 4),
        "Semantic Top Result": semantic_result["document"]
    })
analysis_df = pd.DataFrame(analysis_results)
pd.set_option("display.max_colwidth", 120)
analysis_df
```

```python
for i, row in analysis_df.iterrows():
    print(f"\n{'=' * 100}")
    print(f"QUERY {i + 1}: {row['Query']}")
    print(f"\nKeyword Search (score={row['Keyword Score']})")
    print(row["Keyword Top Result"])
    print(f"\nSemantic Search (distance={row['Semantic Distance']})")
    print(row["Semantic Top Result"])
```

```python
final_results = []
for query in test_queries:
    keyword_results = keyword_search(query, corpus, top_k=3)
    semantic_results = semantic_search(query, top_k=3)
    for rank in range(3):
        final_results.append({
            "Query": query,
            "Rank": rank + 1,
            "Keyword Score": keyword_results[rank]["score"],
            "Keyword Result": keyword_results[rank]["document"],
            "Semantic Distance": round(semantic_results[rank]["distance"], 4),
            "Semantic Result": semantic_results[rank]["document"]
        })
final_df = pd.DataFrame(final_results)
final_df
```
