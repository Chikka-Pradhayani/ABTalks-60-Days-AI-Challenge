```python
import nltk
import re
import string
from nltk.tokenize import word_tokenize,sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
samples = {
    "Social Media": """
OMG!!! I LOVE this movie 😍😍 It's sooo good!!!
Can't stop watching it 😂 #Amazing
""",

    "News Article": """
The Government announced new AI regulations on Monday.
The new rules will affect technology companies and researchers.
""",

    "Customer Review": """
I bought this phone last week!!! The camera is AMAZING 📸,
but the battery isn't very good. I'm disappointed.
"""
}
for name, text in samples.items():
  print("=" * 60)
  print(name)
  print("=" * 60)
  print(text)
```

```python
stop_words = set(stopwords.words('english'))

for name, text in samples.items():

    words = word_tokenize(text.lower())

    before_vocab = set(words)

    filtered_words = [
        word for word in words
        if word not in stop_words
    ]

    after_vocab = set(filtered_words)

    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    print("Before Stopword Removal:")
    print(words)

    print("\nAfter Stopword Removal:")
    print(filtered_words)

    print("\nVocabulary Size Before:", len(before_vocab))
    print("Vocabulary Size After:", len(after_vocab))
```

```python
for name, text in samples.items():

    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    print("BEFORE:")
    print(text)

    cleaned_text = re.sub(r'[^\w\s]', '', text)

    print("\nAFTER:")
    print(cleaned_text)
```

```python
stemmer = PorterStemmer()

for name, text in samples.items():

    words = word_tokenize(text.lower())

    stemmed_words = [
        stemmer.stem(word)
        for word in words
        if word.isalpha()
    ]

    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    print("Original Words:")
    print(words)

    print("\nStemmed Words:")
    print(stemmed_words)
```

```python
lemmatizer = WordNetLemmatizer()

for name, text in samples.items():

    words = word_tokenize(text.lower())

    lemmatized_words = [
        lemmatizer.lemmatize(word)
        for word in words
        if word.isalpha()
    ]

    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    print("Original Words:")
    print(words)

    print("\nLemmatized Words:")
    print(lemmatized_words)
```

```python
for name, text in samples.items():

    words = word_tokenize(text.lower())

    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)

    print(f"{'Original':<20} {'Stemmed':<20} {'Lemmatized':<20}")
    print("-" * 70)

    for word in words:
        if word.isalpha():

            stemmed = stemmer.stem(word)
            lemma = lemmatizer.lemmatize(word)

            print(f"{word:<20} {stemmed:<20} {lemma:<20}")
```

```python
def preprocess_text(text):

    # Convert all characters to lowercase for consistent processing
    text = text.lower()

    # Remove punctuation using regular expressions
    text = re.sub(r'[^\w\s]', '', text)

    # Tokenize the cleaned text into words
    tokens = word_tokenize(text)

    # Remove stopwords
    tokens = [
        word for word in tokens
        if word not in stop_words
    ]

    # Apply stemming
    stemmed = [
        stemmer.stem(word)
        for word in tokens
    ]

    # Apply lemmatization
    lemmatized = [
        lemmatizer.lemmatize(word)
        for word in tokens
    ]

    return tokens, stemmed, lemmatized
```

```python
for name, text in samples.items():

    tokens, stemmed, lemmatized = preprocess_text(text)

    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)

    print("\nOriginal Text:")
    print(text)

    print("\nFinal Tokens:")
    print(tokens)

    print("\nStemmed:")
    print(stemmed)

    print("\nLemmatized:")
    print(lemmatized)
```

```python
import pandas as pd

comparison = []

for name, text in samples.items():

    # Step 1: Lowercase
    lowercase = text.lower()

    # Step 2: Tokenization
    tokens = word_tokenize(lowercase)

    # Step 3: Stopword removal
    no_stopwords = [
        word for word in tokens
        if word.isalpha() and word not in stop_words
    ]

    # Step 4: Punctuation removal
    no_punctuation = re.sub(r'[^\w\s]', '', lowercase)

    # Step 5: Stemming
    stemmed = [
        stemmer.stem(word)
        for word in no_stopwords
    ]

    # Step 6: Lemmatization
    lemmatized = [
        lemmatizer.lemmatize(word)
        for word in no_stopwords
    ]

    comparison.append({
        "Sample": name,
        "Original": text.strip(),
        "Lowercase": lowercase.strip(),
        "Tokens": " ".join(tokens),
        "Stopwords Removed": " ".join(no_stopwords),
        "Punctuation Removed": no_punctuation.strip(),
        "Stemmed": " ".join(stemmed),
        "Lemmatized": " ".join(lemmatized)
    })

comparison_df = pd.DataFrame(comparison)

pd.set_option("display.max_colwidth", 200)

comparison_df
```
