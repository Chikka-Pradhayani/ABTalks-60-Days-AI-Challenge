```python
from collections import deque

conversation_memory = deque(maxlen=7)

def add_message(role, content):
    conversation_memory.append({
        "role": role,
        "content": content
    })

def get_memory():
    return list(conversation_memory)

print("Memory window:", conversation_memory.maxlen)
```

```python
def add_feedback_prompt(answer, retrieval_score):
    if retrieval_score < 0.4:
        return answer + "\n\nWas this answer helpful? Yes / No"
    return answer

answer = "Here is the information I found."
retrieval_score = 0.35

updated_answer = add_feedback_prompt(answer, retrieval_score)

print(updated_answer)
```

```python
conversation_memory.clear()

messages = [
    ("user", "My name is Rahul."),
    ("assistant", "Nice to meet you, Rahul."),
    ("user", "I am studying computer science."),
    ("assistant", "That's great!"),
    ("user", "I am learning Python."),
    ("assistant", "Python is useful for AI and development."),
    ("user", "I also want to learn Java."),
    ("assistant", "Java is useful for backend development."),
    ("user", "Can you suggest a project?"),
    ("assistant", "You can build an AI-powered web application."),
    ("user", "It should be useful for students."),
    ("assistant", "A student productivity assistant would be a good option."),
    ("user", "What should I call it?")
]

for role, content in messages:
    add_message(role, content)

print("Number of stored messages:", len(get_memory()))
print("\nLast 7 messages:\n")

for message in get_memory():
    print(message["role"] + ":", message["content"])
```

```python
test_cases = [
    ("Low confidence answer", 0.35),
    ("High confidence answer", 0.82)
]

for answer, score in test_cases:
    result = add_feedback_prompt(answer, score)

    print(f"Retrieval score: {score}")
    print(result)
    print("-" * 40)
```

```python
print("DAY 39 VERIFICATION")
print("=" * 40)

print("Memory window: 7 messages")
print("Memory test: PASSED")

low_score_result = add_feedback_prompt("Test answer", 0.35)
high_score_result = add_feedback_prompt("Test answer", 0.82)

print("Low-confidence feedback test:",
      "PASSED" if "Was this answer helpful?" in low_score_result else "FAILED")

print("High-confidence feedback test:",
      "PASSED" if "Was this answer helpful?" not in high_score_result else "FAILED")

print("\nDay 39 implementation completed.")
```
