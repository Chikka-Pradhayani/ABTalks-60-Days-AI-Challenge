```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from sentence_transformers import SentenceTransformer
import numpy as np
import pandas as pd
```

```python
document_text = """
Artificial intelligence is one of the most important areas of modern computer science.
It combines computer systems, algorithms, data, and mathematical techniques to create
systems capable of performing tasks that normally require human intelligence.

Machine learning is a major branch of artificial intelligence. Instead of explicitly
programming every rule, machine learning systems learn patterns from data. These systems
can be trained using supervised learning, unsupervised learning, and reinforcement
learning techniques.

Deep learning is a specialized area of machine learning based on artificial neural
networks with multiple layers. Deep learning has achieved significant results in image
recognition, natural language processing, speech recognition, recommendation systems,
and many other applications.

Natural language processing focuses on enabling computers to understand and process
human language. Modern NLP systems use large datasets and neural network architectures
to analyze text, generate responses, summarize documents, translate languages, and
answer questions.

Large language models have changed the way NLP applications are developed. These models
are trained on large collections of text and learn statistical relationships between
words, phrases, and concepts. They can subsequently be adapted for tasks such as
question answering, summarization, classification, and information retrieval.

Information retrieval is the process of finding relevant information from a large
collection of documents. Search engines, recommendation systems, and question-answering
systems commonly use information retrieval techniques.

A major challenge in retrieval augmented generation systems is handling documents that
are too large to fit into the context window of a language model. Chunking addresses
this problem by dividing a document into smaller pieces that can be independently
processed and retrieved.

Chunking strategies have a significant effect on retrieval quality. If chunks are too
small, important context may be lost because related information is separated into
different chunks. If chunks are too large, retrieval may return unnecessary information
and consume more context window space.

Overlap between neighboring chunks can reduce the problem of losing information at
chunk boundaries. When a sentence or concept crosses a boundary, overlapping regions
allow the same information to appear in more than one chunk.

Fixed-size chunking is one of the simplest chunking strategies. It divides text into
pieces according to a fixed number of characters or tokens. Although simple, fixed-size
chunking may split sentences and paragraphs in unnatural locations.

Recursive character splitting provides a more context-aware approach. It attempts to
split text using separators such as paragraphs, newlines, sentences, and spaces before
falling back to smaller units.

The choice of chunk size depends on the document and the retrieval task. Technical
documents may require larger chunks because concepts are often explained across several
sentences. Short articles may work better with smaller chunks.

Retrieval experiments can be used to compare different chunking configurations.
A fixed query can be executed against every configuration and the retrieved results
can then be evaluated for relevance and contextual completeness.

A useful chunking experiment should measure not only whether the correct information
is retrieved but also whether the retrieved chunk contains enough surrounding context
to understand that information.

Therefore, chunk size and overlap should be treated as important parameters in a
retrieval pipeline. Testing multiple configurations on the same document provides
evidence for selecting an appropriate strategy.

""" * 20

print("Document loaded successfully!")
print("Number of characters:", len(document_text))
```

```python
def fixed_size_chunking(text, chunk_size):
    """
    Split text into fixed-size chunks without overlap.
    """
    chunks = []
    for start in range(0, len(text), chunk_size):
        chunk = text[start:start + chunk_size]
        chunks.append(chunk)
    return chunks
```

```python
chunks_100 = fixed_size_chunking(document_text, 100)
print("Number of chunks:", len(chunks_100))
print("\nFirst chunk:")
print(chunks_100[0])
print("\nSecond chunk:")
print(chunks_100[1])
```

```python
def fixed_size_chunking_with_overlap(text, chunk_size, chunk_overlap):
    """
    Split text into fixed-size chunks with configurable overlap.
    """

    if chunk_overlap >= chunk_size:
        raise ValueError("chunk_overlap must be smaller than chunk_size")

    chunks = []
    start = 0

    while start < len(text):
        end = start + chunk_size
        chunk = text[start:end]

        if chunk:
            chunks.append(chunk)

        start += chunk_size - chunk_overlap

    return chunks
```

```python
chunk_sizes = [100, 300, 500, 1000]
overlaps = [0, 50, 100]
all_chunks = {}
for chunk_size in chunk_sizes:
    for overlap in overlaps:
        if overlap >= chunk_size:
            continue
        splitter = RecursiveCharacterTextSplitter(
            chunk_size=chunk_size,
            chunk_overlap=overlap
        )
        chunks_config = splitter.split_text(document_text)
        config_name = f"size={chunk_size}, overlap={overlap}"
        all_chunks[config_name] = chunks_config
print("Total configurations:", len(all_chunks))
```

```python
from sentence_transformers import SentenceTransformer
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")
print("Embedding model loaded successfully!")
```

```python
query = "How does chunk size and overlap affect retrieval quality in document processing?"
print("Query:", query)
```

```python
query_embedding = embedding_model.encode(
    query,
    convert_to_numpy=True
)
print("Query embedding shape:", query_embedding.shape)
```

```python
from numpy.linalg import norm
def cosine_similarity(vector_a, vector_b):
    return np.dot(vector_a, vector_b) / (
        norm(vector_a) * norm(vector_b)
    )
```

```python
test_chunks = all_chunks["size=300, overlap=50"]
test_embeddings = embedding_model.encode(
    test_chunks,
    convert_to_numpy=True
)
scores = []
for i, chunk_embedding in enumerate(test_embeddings):
    score = cosine_similarity(
        query_embedding,
        chunk_embedding
    )
    scores.append((i, score))
```

```python
scores.sort(key=lambda x: x[1], reverse=True)
top_3 = scores[:3]
for rank, (index, score) in enumerate(top_3, start=1):
    print(f"\n{'='*60}")
    print(f"Rank {rank}")
    print(f"Similarity Score: {score:.4f}")
    print(f"Chunk Index: {index}")
    print(f"{'='*60}")
    print(test_chunks[index])
```

```python
def retrieve_top_chunks(chunks, query_embedding, top_k=3):
    """
    Retrieve the top-k most similar chunks for a query.
    """
    chunk_embeddings = embedding_model.encode(
        chunks,
        convert_to_numpy=True
    )
    scores = []
    for i, chunk_embedding in enumerate(chunk_embeddings):
        score = cosine_similarity(
            query_embedding,
            chunk_embedding
        )
        scores.append((i, score))
    scores.sort(key=lambda x: x[1], reverse=True)
    return scores[:top_k]
```

```python
retrieval_results = {}
for config, chunks_list in all_chunks.items():
    top_chunks = retrieve_top_chunks(
        chunks_list,
        query_embedding,
        top_k=3
    )
    retrieval_results[config] = top_chunks
print("Retrieval completed for all configurations!")
```

```python
for config, results in retrieval_results.items():
    print("\n" + "=" * 80)
    print(config)
    print("=" * 80)
    chunks_list = all_chunks[config]
    for rank, (index, score) in enumerate(results, start=1):
        print(f"\nRank {rank} | Score: {score:.4f}")
        print("-" * 60)
        print(chunks_list[index])
```

```python
comparison_data = []
for config, results in retrieval_results.items():
    size = int(config.split(",")[0].split("=")[1])
    overlap = int(config.split(",")[1].split("=")[1])
    scores = [score for _, score in results]
    comparison_data.append({
        "Chunk Size": size,
        "Overlap": overlap,
        "Top 1 Score": scores[0],
        "Top 2 Score": scores[1],
        "Top 3 Score": scores[2],
        "Average Score": np.mean(scores)
    })
comparison_df = pd.DataFrame(comparison_data)
comparison_df = comparison_df.sort_values(
    "Average Score",
    ascending=False
).reset_index(drop=True)
comparison_df
```

```python
def find_boundary_failures(text, chunk_size, chunk_overlap=0, max_examples=3):
    """
    Find cases where a chunk boundary occurs inside a sentence.
    """
    failures = []
    step = chunk_size - chunk_overlap
    start = 0
    while start < len(text) and len(failures) < max_examples:
        end = min(start + chunk_size, len(text))
        if end >= len(text):
            break
        before = text[end - 1]
        after = text[end]
        punctuation = ".!?\n"
        if before not in punctuation:
            sentence_start = text.rfind(".", 0, end)
            sentence_start = max(
                sentence_start,
                text.rfind("!", 0, end),
                text.rfind("?", 0, end)
            )
            sentence_start += 1
            sentence_end_candidates = [
                text.find(".", end),
                text.find("!", end),
                text.find("?", end)
            ]
            valid_candidates = [
                position
                for position in sentence_end_candidates
                if position != -1
            ]
            if valid_candidates:
                sentence_end = min(valid_candidates) + 1
                failures.append({
                    "boundary_position": end,
                    "sentence_start": sentence_start,
                    "sentence_end": sentence_end,
                    "sentence": text[sentence_start:sentence_end]
                })
        start += step
    return failures
```

```python
failures = find_boundary_failures(
    document_text,
    chunk_size=300,
    chunk_overlap=0,
    max_examples=3
)
print("Number of boundary failures found:", len(failures))
```
