```python
from datetime import date
```

```python
documents = [
    {
        "id": 1,
        "text": "A document retrieval system finds relevant documents based on a user's query."
    },
    {
        "id": 2,
        "text": "TF-IDF represents documents using numerical vectors based on word importance."
    },
    {
        "id": 3,
        "text": "Cosine similarity measures how similar two vectors are."
    },
    {
        "id": 4,
        "text": "A typical retrieval pipeline contains preprocessing, vectorization, similarity calculation, and retrieval."
    },
    {
        "id": 5,
        "text": "Our AI learning course contains 5 modules, and each module contains 8 lessons."
    }
]
```

```python
def calculator(expression):
    try:
        result = eval(expression)
        return str(result)
    except Exception as e:
        return f"Calculator error: {e}"
```

```python
print(calculator("25 * 4"))
print(calculator("100 / 5"))
print(calculator("10 + 20"))
```

```python
def search_docs(query):
    query_words = query.lower().split()

    results = []

    for doc in documents:
        text = doc["text"].lower()

        score = sum(word in text for word in query_words)

        if score > 0:
            results.append((score, doc["text"]))

    results.sort(reverse=True)

    if not results:
        return "No relevant documents found."

    return "\n".join(
        f"- {text}"
        for score, text in results[:3]
    )
```

```python
print(search_docs("modules lessons"))
```

```python
def get_today():
    return str(date.today())
print(get_today())
```

```python
tool_registry = {
    "calculator": calculator,
    "search_docs": search_docs,
    "get_today": get_today
}

print(tool_registry)
```

```python
def dispatch(tool_name, parameters):
    if tool_name not in tool_registry:
        return f"Error: Unknown tool '{tool_name}'"

    tool = tool_registry[tool_name]

    try:
        return tool(parameters)
    except Exception as e:
        return f"Tool error: {e}"
print(dispatch("calculator", "25 * 4"))
```

```python
print(dispatch("search_docs", "modules lessons"))
print(dispatch("get_today", ""))
```

```python
def run_agent(agent):
    while True:

        response = agent.next_action()

        if response["type"] == "final":
            print("\nFinal Answer:")
            print(response["answer"])
            break

        print("\nThought:")
        print(response["thought"])

        print("\nAction:")
        print(f"{response['tool']}({response['input']!r})")

        observation = dispatch(
            response["tool"],
            response["input"]
        )

        print("\nObservation:")
        print(observation)
```

```python
class MockLLM:
    def __init__(self, plan):
        self.plan = plan
        self.step = 0

    def next_action(self):
        if self.step >= len(self.plan):
            return {
                "type": "final",
                "answer": "Task completed."
            }

        action = self.plan[self.step]
        self.step += 1

        return action
```

```python
plan = [
    {
        "type": "action",
        "thought": "I need to find the number of modules and lessons from the documents.",
        "tool": "search_docs",
        "input": "modules lessons"
    },
    {
        "type": "action",
        "thought": "The document says there are 5 modules and 8 lessons per module. I need to calculate the total.",
        "tool": "calculator",
        "input": "5 * 8"
    },
    {
        "type": "final",
        "answer": "There are 40 lessons in total."
    }
]

agent = MockLLM(plan)
```

```python
run_agent(agent)
```
