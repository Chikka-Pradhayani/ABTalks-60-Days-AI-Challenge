```python
from datetime import date


# Tool 1: Search documents
def search_documents(query):
    # Stub document database for now
    documents = {
        "python": "Python is a high-level programming language known for simplicity and readability.",
        "rag": "RAG stands for Retrieval-Augmented Generation. It retrieves relevant documents before generating an answer.",
        "fastapi": "FastAPI is a Python framework used for building APIs quickly.",
        "machine learning": "Machine learning allows computers to learn patterns from data."
    }

    query_lower = query.lower()

    for keyword, content in documents.items():
        if keyword in query_lower:
            return content

    return "No relevant document was found."


# Tool 2: Weather stub
def get_weather_stub(city):
    weather_data = {
        "hyderabad": "Sunny, 30°C",
        "delhi": "Cloudy, 28°C",
        "mumbai": "Rainy, 27°C",
        "bangalore": "Partly cloudy, 25°C"
    }

    return weather_data.get(
        city.lower(),
        f"Weather data for {city} is not available."
    )


# Tool 3: Calculator
def calculate(expression):
    try:
        # Basic calculator for testing
        result = eval(expression, {"__builtins__": {}}, {})
        return str(result)

    except Exception as e:
        return f"Calculation error: {str(e)}"


# Tool 4: Today's date
def get_today():
    return str(date.today())


print("All four tools are ready!")
```

```python
print(search_documents("What is RAG?"))

print(get_weather_stub("Hyderabad"))

print(calculate("25 * 4 + 10"))

print(get_today())
```

```python
def execute_tool(tool_name, arguments):

    if tool_name not in tool_functions:
        return {
            "error": f"Unknown tool: {tool_name}"
        }

    try:
        function = tool_functions[tool_name]

        result = function(**arguments)

        return {
            "tool": tool_name,
            "arguments": arguments,
            "result": result
        }

    except Exception as e:
        return {
            "tool": tool_name,
            "arguments": arguments,
            "error": str(e)
        }
result = execute_tool(
    "calculate",
    {"expression": "100 / 4"}
)

print(result)
```

```python
def validate_tool_arguments(tool_name, arguments):

    if not isinstance(arguments, dict):
        return False, "Arguments must be a dictionary"

    if tool_name == "search_documents":

        if "query" not in arguments:
            return False, "Missing required argument: query"

        if not isinstance(arguments["query"], str):
            return False, "query must be a string"

    elif tool_name == "get_weather_stub":

        if "city" not in arguments:
            return False, "Missing required argument: city"

        if not isinstance(arguments["city"], str):
            return False, "city must be a string"

    elif tool_name == "calculate":

        if "expression" not in arguments:
            return False, "Missing required argument: expression"

        if not isinstance(arguments["expression"], str):
            return False, "expression must be a string"

    elif tool_name == "get_today":

        if len(arguments) != 0:
            return False, "get_today does not accept arguments"

    else:
        return False, f"Unknown tool: {tool_name}"

    return True, "Arguments are valid"


print("Validation function created.")
```

```python
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI(title="Structured Tool Calling Agent")


class AgentRequest(BaseModel):
    question: str


@app.post("/agent")
def agent(request: AgentRequest):

    question = request.question
    question_lower = question.lower()

    tool_calls = []
    results = []

    # Tool selection logic for testing
    if "weather" in question_lower:
        city = "Hyderabad"

        call = {
            "tool": "get_weather_stub",
            "arguments": {
                "city": city
            }
        }

        result = execute_tool(
            call["tool"],
            call["arguments"]
        )

        tool_calls.append(call)
        results.append(result)

        final_answer = f"The weather in {city} is {result['result']}."

    elif "today" in question_lower or "date" in question_lower:

        call = {
            "tool": "get_today",
            "arguments": {}
        }

        result = execute_tool(
            call["tool"],
            call["arguments"]
        )

        tool_calls.append(call)
        results.append(result)

        final_answer = f"Today's date is {result['result']}."

    elif "calculate" in question_lower:

        expression = question_lower.replace("calculate", "").strip()

        call = {
            "tool": "calculate",
            "arguments": {
                "expression": expression
            }
        }

        result = execute_tool(
            call["tool"],
            call["arguments"]
        )

        tool_calls.append(call)
        results.append(result)

        final_answer = f"The result is {result['result']}."

    elif "rag" in question_lower:

        call = {
            "tool": "search_documents",
            "arguments": {
                "query": question
            }
        }

        result = execute_tool(
            call["tool"],
            call["arguments"]
        )

        tool_calls.append(call)
        results.append(result)

        final_answer = result["result"]

    else:
        final_answer = "I could not determine which tool to use."

    return {
        "question": question,
        "final_answer": final_answer,
        "tool_calls": tool_calls,
        "tool_results": results
    }


print("FastAPI /agent endpoint created!")
```

```python
test_request = AgentRequest(
    question="What is the weather in Hyderabad?"
)

response = agent(test_request)

print(response)
```
