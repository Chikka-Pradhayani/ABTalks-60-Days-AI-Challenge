```python
documents = [
    "Machine learning enables computers to learn patterns from data.",
    "Deep learning uses neural networks with many layers.",
    "Natural language processing helps computers understand human language.",
    "Computer vision allows machines to interpret images and videos.",
    "Generative AI can create text, images, audio, and code.",
    "Large language models are trained on massive text datasets.",
    "Vector databases store numerical representations of information.",
    "Embeddings represent text as vectors in a mathematical space.",
    "Semantic search finds information based on meaning rather than exact words.",
    "RAG combines information retrieval with generative AI.",
    "Recommendation systems suggest products based on user preferences.",
    "Data preprocessing improves the quality of machine learning models.",
    "Supervised learning uses labeled training examples.",
    "Unsupervised learning discovers patterns in unlabeled data.",
    "Reinforcement learning trains agents using rewards and penalties.",
    "Neural networks are inspired by the structure of biological brains.",
    "AI assistants use language models to answer user questions.",
    "Fine-tuning adapts a pretrained model to a specific task.",
    "Model evaluation measures how well an AI system performs.",
    "MLOps helps organizations deploy and maintain machine learning models."
]
```

```python
print("Number of documents:", len(documents))
```

```python
from sentence_transformers import SentenceTransformer

# Load the embedding model
model = SentenceTransformer("all-MiniLM-L6-v2")

# Convert all 20 sentences into embeddings
embeddings = model.encode(documents)

print("Number of embeddings:", len(embeddings))
print("Embedding shape:", embeddings.shape)
```

```python
print(embeddings[0])
```

```python
query = input("Enter your search query: ")

print("Your query:", query)
```

```python
query_embedding = model.encode(query)

print("Query embedding shape:", query_embedding.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# Calculate similarity between query and all documents
similarities = cosine_similarity(
    [query_embedding],
    embeddings
)[0]

print(similarities)
```

```python
# Get the indices of the top 3 most similar sentences
top_3_indices = similarities.argsort()[-3:][::-1]

# Display the top 3 results
print("Top 3 Semantic Search Results:\n")

for i, index in enumerate(top_3_indices, start=1):
    print(f"{i}. {documents[index]}")
    print(f"   Similarity Score: {similarities[index]:.4f}\n")
```

```python
def keyword_search(query, documents, top_k=3):
    query_words = set(query.lower().split())

    results = []

    for i, document in enumerate(documents):
        document_words = set(document.lower().split())

        # Count matching words
        score = len(query_words.intersection(document_words))

        results.append((score, i))

    # Sort by keyword score
    results.sort(reverse=True)

    return results[:top_k]
```

```python
keyword_results = keyword_search(query, documents)

print("Top 3 Keyword Search Results:\n")

for rank, (score, index) in enumerate(keyword_results, start=1):
    print(f"{rank}. {documents[index]}")
    print(f"   Keyword Score: {score}\n")
```

```python
print("QUERY:")
print(query)

print("\n" + "=" * 60)
print("SEMANTIC SEARCH — TOP 3")
print("=" * 60)

for rank, index in enumerate(top_3_indices, start=1):
    print(f"{rank}. {documents[index]}")
    print(f"   Similarity Score: {similarities[index]:.4f}")

print("\n" + "=" * 60)
print("KEYWORD SEARCH — TOP 3")
print("=" * 60)

for rank, (score, index) in enumerate(keyword_results, start=1):
    print(f"{rank}. {documents[index]}")
    print(f"   Keyword Score: {score}")
```

```python
test_queries = [
    "How can computers understand pictures?",
    "How can AI find information based on meaning?",
    "How do machines learn from examples?"
]

for query in test_queries:
    print("\n" + "=" * 70)
    print("QUERY:", query)
    print("=" * 70)

    # Convert query to embedding
    query_embedding = model.encode(query)

    # Calculate semantic similarity
    similarities = cosine_similarity(
        [query_embedding],
        embeddings
    )[0]

    # Get top 3 semantic results
    top_3_indices = similarities.argsort()[-3:][::-1]

    print("\nTop 3 Semantic Results:")

    for rank, index in enumerate(top_3_indices, start=1):
        print(
            f"{rank}. {documents[index]} "
            f"(Score: {similarities[index]:.4f})"
        )
```
