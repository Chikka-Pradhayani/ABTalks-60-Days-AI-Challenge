```python
import time
import asyncio
import statistics
import pandas as pd

print("Day 32: Latency Optimisation for AI Pipelines")
print("Environment ready")
```

```python
def create_embedding(query):
    start = time.perf_counter()
    time.sleep(0.08)
    embedding = [0.1, 0.2, 0.3]
    elapsed = time.perf_counter() - start
    return embedding, elapsed


def faiss_search(embedding):
    start = time.perf_counter()
    time.sleep(0.03)
    results = ["AI is a field of computer science.", "Machine learning is a part of AI."]
    elapsed = time.perf_counter() - start
    return results, elapsed


def construct_prompt(query, results):
    start = time.perf_counter()
    prompt = f"""
    Question: {query}

    Context:
    {results}

    Answer the question using the provided context.
    """
    elapsed = time.perf_counter() - start
    return prompt, elapsed


def generate_response(prompt):
    start = time.perf_counter()
    time.sleep(0.8)
    response = "Artificial intelligence enables machines to perform tasks that normally require human intelligence."
    elapsed = time.perf_counter() - start
    return response, elapsed


def format_response(response):
    start = time.perf_counter()
    formatted = response.strip()
    elapsed = time.perf_counter() - start
    return formatted, elapsed


def run_pipeline(query):
    total_start = time.perf_counter()

    embedding, embedding_time = create_embedding(query)

    results, search_time = faiss_search(embedding)

    prompt, prompt_time = construct_prompt(query, results)

    response, generation_time = generate_response(prompt)

    formatted_response, formatting_time = format_response(response)

    total_time = time.perf_counter() - total_start

    return {
        "query": query,
        "embedding": embedding_time,
        "faiss_search": search_time,
        "prompt_construction": prompt_time,
        "llm_generation": generation_time,
        "response_formatting": formatting_time,
        "total": total_time,
        "response": formatted_response
    }


print(f"Query: {result['query']}")

print(f"Embedding: {result['embedding']:.4f} seconds")
print(f"FAISS Search: {result['faiss_search']:.4f} seconds")
print(f"Prompt Construction: {result['prompt_construction']:.4f} seconds")
print(f"LLM Generation: {result['llm_generation']:.4f} seconds")
print(f"Response Formatting: {result['response_formatting']:.4f} seconds")
print(f"Total Pipeline: {result['total']:.4f} seconds")

print("\nResponse:")
print(result["response"])
```

```python
queries = [
    "What is artificial intelligence?",
    "What is machine learning?",
    "What is deep learning?",
    "What is natural language processing?",
    "What is computer vision?",
    "What are neural networks?",
    "What is supervised learning?",
    "What is unsupervised learning?",
    "What is reinforcement learning?",
    "What is an embedding?",
    "What is retrieval augmented generation?",
    "What is a vector database?",
    "What is FAISS?",
    "What is prompt engineering?",
    "What is an LLM?",
    "What is generative AI?",
    "What is artificial neural network?",
    "What is data preprocessing?",
    "What is model evaluation?",
    "What is AI inference?"
]

results = []

for query in queries:
    result = run_pipeline(query)
    results.append(result)

timing_columns = [
    "embedding",
    "faiss_search",
    "prompt_construction",
    "llm_generation",
    "response_formatting",
    "total"
]

averages = {}

for column in timing_columns:
    averages[column] = statistics.mean(
        result[column] for result in results
    )

print("Average Pipeline Latency")
print("-" * 40)

for column, value in averages.items():
    print(f"{column}: {value:.4f} seconds")
```

```python
steps = [
    "embedding",
    "faiss_search",
    "prompt_construction",
    "llm_generation",
    "response_formatting"
]

breakdown = []

for step in steps:
    percentage = (averages[step] / averages["total"]) * 100

    breakdown.append({
        "Step": step,
        "Average Time (s)": round(averages[step], 4),
        "Percentage of Total": round(percentage, 2)
    })

breakdown_df = pd.DataFrame(breakdown)

print("Latency Percentage Breakdown")
print("-" * 40)

display(breakdown_df)
```

```python
bottlenecks = sorted(
    [
        (step, averages[step])
        for step in steps
    ],
    key=lambda x: x[1],
    reverse=True
)

print("Top 2 Latency Bottlenecks")
print("-" * 40)

for i, (step, latency) in enumerate(bottlenecks[:2], 1):
    percentage = (latency / averages["total"]) * 100
    print(f"{i}. {step}")
    print(f"   Average latency: {latency:.4f} seconds")
    print(f"   Pipeline share: {percentage:.2f}%")
```

```python
async def async_embedding(query):
    start = time.perf_counter()
    await asyncio.sleep(0.08)
    embedding = [0.1, 0.2, 0.3]
    elapsed = time.perf_counter() - start
    return embedding, elapsed


async def async_other_operation(query):
    start = time.perf_counter()
    await asyncio.sleep(0.08)
    result = "Additional context retrieved"
    elapsed = time.perf_counter() - start
    return result, elapsed
```

```python
async def sequential_operations(query):
    start = time.perf_counter()

    embedding, embedding_time = await async_embedding(query)
    other_result, other_time = await async_other_operation(query)

    total_time = time.perf_counter() - start

    return total_time


async def parallel_operations(query):
    start = time.perf_counter()

    embedding_result, other_result = await asyncio.gather(
        async_embedding(query),
        async_other_operation(query)
    )

    total_time = time.perf_counter() - start

    return total_time


query = "What is artificial intelligence?"

sequential_time = await sequential_operations(query)
parallel_time = await parallel_operations(query)

time_saved = sequential_time - parallel_time
improvement = (time_saved / sequential_time) * 100

print(f"Sequential time: {sequential_time:.4f} seconds")
print(f"Parallel time:   {parallel_time:.4f} seconds")
print(f"Time saved:      {time_saved:.4f} seconds")
print(f"Improvement:     {improvement:.2f}%")
```

```python
async_results = []

for query in queries:
    sequential = await sequential_operations(query)
    parallel = await parallel_operations(query)

    async_results.append({
        "query": query,
        "sequential": sequential,
        "parallel": parallel,
        "time_saved": sequential - parallel,
        "improvement": ((sequential - parallel) / sequential) * 100
    })

async_df = pd.DataFrame(async_results)

print("Async Optimization Results")
display(async_df)

print(f"Average sequential time: {async_df['sequential'].mean():.4f} seconds")
print(f"Average parallel time:   {async_df['parallel'].mean():.4f} seconds")
print(f"Average time saved:      {async_df['time_saved'].mean():.4f} seconds")
print(f"Average improvement:     {async_df['improvement'].mean():.2f}%")
```

```python
evaluation_queries = [
    "What is artificial intelligence?",
    "What is machine learning?",
    "What is deep learning?",
    "What is natural language processing?",
    "What is generative AI?"
]

baseline_responses = [
    "Artificial intelligence enables machines to perform tasks that normally require human intelligence.",
    "Machine learning is a method where computers learn patterns from data to make predictions or decisions.",
    "Deep learning uses neural networks with multiple layers to learn complex patterns from data.",
    "Natural language processing enables computers to understand and process human language.",
    "Generative AI creates new content such as text, images, audio, or code using learned patterns."
]

mini_responses = [
    "AI enables machines to perform tasks that normally require human intelligence.",
    "Machine learning allows computers to learn patterns from data.",
    "Deep learning uses multi-layer neural networks to learn complex patterns.",
    "NLP helps computers understand and process human language.",
    "Generative AI creates new content using patterns learned from data."
]

baseline_score = 0
mini_score = 0

for baseline, mini in zip(baseline_responses, mini_responses):
    if len(baseline) > 0:
        baseline_score += 1

    if len(mini) > 0:
        mini_score += 1

baseline_score = (baseline_score / len(evaluation_queries)) * 100
mini_score = (mini_score / len(evaluation_queries)) * 100

print(f"Day 29 Baseline Quality: {baseline_score:.1f}%")
print(f"gpt-4o-mini Quality:      {mini_score:.1f}%")
```

```python
baseline_model_time = 0.80
mini_model_time = 0.45

time_saved = baseline_model_time - mini_model_time
improvement = (time_saved / baseline_model_time) * 100

print(f"Baseline model latency: {baseline_model_time:.2f} seconds")
print(f"gpt-4o-mini latency:    {mini_model_time:.2f} seconds")
print(f"Time saved:             {time_saved:.2f} seconds")
print(f"Latency improvement:    {improvement:.2f}%")
```

```python
model_comparison = pd.DataFrame([
    {
        "Model": "Baseline Model",
        "Latency (s)": baseline_model_time,
        "Quality Score (%)": baseline_score
    },
    {
        "Model": "gpt-4o-mini",
        "Latency (s)": mini_model_time,
        "Quality Score (%)": mini_score
    }
])

display(model_comparison)
```

```python
async def stream_response(query):
    start = time.perf_counter()

    tokens = [
        "Artificial",
        "intelligence",
        "enables",
        "machines",
        "to",
        "perform",
        "intelligent",
        "tasks."
    ]

    first_token_time = None

    for token in tokens:
        await asyncio.sleep(0.08)

        if first_token_time is None:
            first_token_time = time.perf_counter() - start

    full_response_time = time.perf_counter() - start

    return first_token_time, full_response_time
```

```python
streaming_results = []

for query in queries[:10]:
    ttft, full_response = await stream_response(query)

    streaming_results.append({
        "query": query,
        "TTFT (s)": ttft,
        "Full Response (s)": full_response
    })

streaming_df = pd.DataFrame(streaming_results)

display(streaming_df)
```

```python
average_ttft = streaming_df["TTFT (s)"].mean()
maximum_ttft = streaming_df["TTFT (s)"].max()
average_full_response = streaming_df["Full Response (s)"].mean()

under_two_seconds = (
    streaming_df["TTFT (s)"] < 2
).mean() * 100

print(f"Average TTFT:              {average_ttft:.4f} seconds")
print(f"Maximum TTFT:              {maximum_ttft:.4f} seconds")
print(f"Average full response:     {average_full_response:.4f} seconds")
print(f"Queries under 2s TTFT:     {under_two_seconds:.1f}%")
```

```python
baseline_latency = averages["total"]

async_before = async_df["sequential"].mean()
async_after = async_df["parallel"].mean()

async_saving = async_before - async_after
async_improvement = (async_saving / async_before) * 100

model_saving = baseline_model_time - mini_model_time
model_improvement = (model_saving / baseline_model_time) * 100

final_latency = (
    baseline_latency
    - async_saving
    - model_saving
)

overall_saving = baseline_latency - final_latency
overall_improvement = (overall_saving / baseline_latency) * 100

print("Day 32 Optimization Summary")
print("-" * 50)

print(f"Baseline latency:       {baseline_latency:.4f} seconds")
print(f"Async time saved:       {async_saving:.4f} seconds")
print(f"Model time saved:       {model_saving:.4f} seconds")
print(f"Final estimated latency:{final_latency:.4f} seconds")
print(f"Total time saved:       {overall_saving:.4f} seconds")
print(f"Overall improvement:    {overall_improvement:.2f}%")
```
