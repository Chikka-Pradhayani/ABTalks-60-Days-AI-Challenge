```python
from dataclasses import dataclass, field
from typing import List
```

```python
@dataclass
class WorkflowState:
    topic: str
    retrieved_chunks: List[str] = field(default_factory=list)
    extracted_points: List[str] = field(default_factory=list)
    synthesis: str = ""
    final_report: str = ""
```

```python
def save_checkpoint(state, step_name):
    filename = f"{step_name}.json"

    with open(filename, "w") as file:
        json.dump(state.__dict__, file, indent=4)

    print(f"Checkpoint saved: {filename}")
```

```python
def run_workflow(topic):
    state = search_sources(topic)
    save_checkpoint(state, "search_sources")

    state.extracted_points = extract_key_points(
        state.retrieved_chunks
    )
    save_checkpoint(state, "extract_key_points")

    state.synthesis = synthesise_findings(
        state.extracted_points
    )
    save_checkpoint(state, "synthesise_findings")

    state.final_report = format_report(
        state.synthesis
    )
    save_checkpoint(state, "format_report")

    return state
```

```python
def load_checkpoint(step_name):
    filename = f"{step_name}.json"
    try:
        with open(filename, "r") as file:
            data = json.load(file)
        return WorkflowState(**data)
    except FileNotFoundError:
        return None
```

```python
def run_workflow(topic):
    state = load_checkpoint("search_sources")
    if state is None:
        state = search_sources(topic)
        save_checkpoint(state, "search_sources")
    state = load_checkpoint("extract_key_points") or state
    if not state.extracted_points:
        state.extracted_points = extract_key_points(state.retrieved_chunks)
        save_checkpoint(state, "extract_key_points")
    state = load_checkpoint("synthesise_findings") or state
    if not state.synthesis:
        state.synthesis = synthesise_findings(state.extracted_points)
        save_checkpoint(state, "synthesise_findings")
    state = load_checkpoint("format_report") or state
    if not state.final_report:
        state.final_report = format_report(state.synthesis)
        save_checkpoint(state, "format_report")
    return state
```

```python
def handle_error(step_name, error, state):
    print(f"Error in {step_name}: {error}")
    print("Current state:", state)
    save_checkpoint(state, f"partial_{step_name}")
```
