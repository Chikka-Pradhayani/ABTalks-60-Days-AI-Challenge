```python
def retrieve_documents(query, top_k=3, title_filter=None):

    documents_to_search = langchain_documents

    if title_filter:
        documents_to_search = [
            doc for doc in langchain_documents
            if doc.metadata.get("title") == title_filter
        ]

    if not documents_to_search:
        return []

    temp_vectorstore = FAISS.from_documents(
        documents_to_search,
        embeddings
    )

    results = temp_vectorstore.similarity_search_with_score(
        query,
        k=top_k
    )

    retrieved_docs = []

    for doc, score in results:
        retrieved_docs.append({
            "id": doc.metadata.get("id"),
            "title": doc.metadata.get("title"),
            "content": doc.page_content,
            "score": float(score)
        })

    return retrieved_docs
```

```python
import time

from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.documents import Document
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
from transformers import pipeline

documents = [
    {
        "id": "doc_1",
        "title": "Artificial Intelligence",
        "content": "Artificial Intelligence is the field of computer science that focuses on creating systems capable of performing tasks that normally require human intelligence."
    },
    {
        "id": "doc_2",
        "title": "Machine Learning",
        "content": "Machine Learning is a subset of Artificial Intelligence that enables computers to learn patterns from data and make predictions or decisions without being explicitly programmed."
    },
    {
        "id": "doc_3",
        "title": "Retrieval Augmented Generation",
        "content": "Retrieval Augmented Generation, or RAG, combines information retrieval with language generation. Relevant documents are retrieved first and then provided to a language model as context."
    }
]

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=300,
    chunk_overlap=50
)

chunks = []

for doc in documents:
    doc_chunks = text_splitter.split_text(doc["content"])

    for i, chunk in enumerate(doc_chunks):
        chunks.append({
            "id": f'{doc["id"]}_chunk_{i}',
            "title": doc["title"],
            "content": chunk
        })

langchain_documents = [
    Document(
        page_content=chunk["content"],
        metadata={
            "id": chunk["id"],
            "title": chunk["title"]
        }
    )
    for chunk in chunks
]

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

vectorstore = FAISS.from_documents(
    langchain_documents,
    embeddings
)

def retrieve_documents(query, top_k=3, title_filter=None):

    documents_to_search = langchain_documents

    if title_filter:
        documents_to_search = [
            doc for doc in langchain_documents
            if doc.metadata["title"].lower() == title_filter.lower()
        ]

    if not documents_to_search:
        return []

    filtered_vectorstore = FAISS.from_documents(
        documents_to_search,
        embeddings
    )

    results = filtered_vectorstore.similarity_search_with_score(
        query,
        k=min(top_k, len(documents_to_search))
    )

    retrieved_docs = []

    for doc, score in results:
        retrieved_docs.append({
            "id": doc.metadata["id"],
            "title": doc.metadata["title"],
            "content": doc.page_content,
            "score": float(score)
        })

    return retrieved_docs


GROUNDING_PROMPT = """
Answer the question using ONLY the retrieved context.

Retrieved Context:
{context}

Question:
{query}

Answer:
"""

generator = pipeline(
    "text-generation",
    model="distilgpt2"
)

def generate_answer(prompt):

    result = generator(
        prompt,
        max_new_tokens=100,
        do_sample=False
    )

    return result[0]["generated_text"].strip()


def run_rag(query, top_k=3, title_filter=None):

    retrieved_docs = retrieve_documents(
        query=query,
        top_k=top_k,
        title_filter=title_filter
    )

    if not retrieved_docs:
        return {
            "answer": "I don't have enough information in the provided knowledge base.",
            "sources": [],
            "confidence": 0.0
        }

    context = "\n\n".join(
        f"Source ID: {doc['id']}\n"
        f"Title: {doc['title']}\n"
        f"Content: {doc['content']}"
        for doc in retrieved_docs
    )

    prompt = GROUNDING_PROMPT.format(
        context=context,
        query=query
    )

    answer = generate_answer(prompt)

    sources = [doc["id"] for doc in retrieved_docs]

    answer += "\n\nSources: " + ", ".join(sources)

    return {
        "answer": answer,
        "sources": sources,
        "confidence": max(
            1 / (1 + doc["score"])
            for doc in retrieved_docs
        )
    }

print("RAG pipeline ready")
```

```python
latency_questions = [
    "What is the main purpose of the system?",
    "What problem does the system solve?",
    "What are the main features?",
    "Who can use this system?",
    "How does the system work?",
    "What technologies are used?",
    "What are the main benefits?",
    "What are the limitations?",
    "How is the system implemented?",
    "Give me a short summary of the system."
]

latency_results = []

for number, question in enumerate(latency_questions, start=1):

    print(f"\nTest {number}: {question}")

    start_time = time.time()
    rag_result = run_rag(question, top_k=3)
    retrieval_time = time.time() - start_time

    start_time = time.time()
    answer = generate_answer(
        GROUNDING_PROMPT.format(
            context="\n\n".join(
                f"Source ID: {doc['id']}\n"
                f"Title: {doc['title']}\n"
                f"Content: {doc['content']}"
                for doc in rag_result.get("retrieved_documents", [])
            ),
            query=question
        )
    )
    generation_time = time.time() - start_time

    total_time = retrieval_time + generation_time

    latency_results.append({
        "question": question,
        "retrieval_time": retrieval_time,
        "generation_time": generation_time,
        "total_time": total_time
    })

    print(f"Retrieval: {retrieval_time:.2f} seconds")
    print(f"Generation: {generation_time:.2f} seconds")
    print(f"Total: {total_time:.2f} seconds")


average_retrieval = sum(
    r["retrieval_time"] for r in latency_results
) / len(latency_results)

average_generation = sum(
    r["generation_time"] for r in latency_results
) / len(latency_results)

average_total = sum(
    r["total_time"] for r in latency_results
) / len(latency_results)

slowest = max(
    latency_results,
    key=lambda r: r["total_time"]
)

print("\nLatency testing completed.")
print(f"Average retrieval time: {average_retrieval:.2f} seconds")
print(f"Average generation time: {average_generation:.2f} seconds")
print(f"Average total time: {average_total:.2f} seconds")
print(f"Slowest query: {slowest['question']}")
print(f"Slowest time: {slowest['total_time']:.2f} seconds")
```

```python
degradation_questions = [
    "What is quantum computing?",
    "Who is the current president of the United States?",
    "How do I cook biryani?",
    "What is the population of Hyderabad?",
    "How does blockchain mining work?"
]

degradation_results = []

for number, question in enumerate(degradation_questions, start=1):

    print(f"\nTest {number}: {question}")

    result = run_rag(question, top_k=3)

    answer = result["answer"]
    confidence = result["confidence"]
    sources = result["sources"]

    print("Answer:")
    print(answer)

    print("\nSources:", sources)
    print(f"Confidence: {confidence:.3f}")

    honest_response = (
        "I don't have enough information"
        in answer
        or "low confidence" in answer.lower()
    )

    degradation_results.append({
        "question": question,
        "answer": answer,
        "sources": sources,
        "confidence": confidence,
        "honest_response": honest_response
    })

passed = sum(
    result["honest_response"]
    for result in degradation_results
)

print("\nGraceful degradation testing completed.")
print(f"Passed: {passed}/5")
print(f"Failed: {5 - passed}/5")
```
