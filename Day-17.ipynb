```python
from langchain_core.documents import Document
print("LangChain Document imported successfully!")
```

```python
langchain_documents = []

for item in documents_with_metadata:

    doc = Document(
        page_content=item["text"],
        metadata=item["metadata"]
    )

    langchain_documents.append(doc)

print("LangChain documents created:", len(langchain_documents))
```

```python
print("CONTENT:")
print(langchain_documents[0].page_content)

print("\nMETADATA:")
print(langchain_documents[0].metadata)
```

```python
metadata_texts = [
    doc.page_content
    for doc in langchain_documents
]

print("Texts ready:", len(metadata_texts))
```

```python
from sentence_transformers import SentenceTransformer

embedder = SentenceTransformer("all-MiniLM-L6-v2")

print("Embedder loaded successfully!")
```

```python
metadata_embeddings = embedder.encode(
    metadata_texts,
    convert_to_numpy=True
).astype("float32")

print("Embedding shape:", metadata_embeddings.shape)
```

```python
import faiss
print("FAISS imported successfully!")
```

```python
metadata_index = faiss.IndexFlatL2(
    metadata_embeddings.shape[1]
)

metadata_index.add(
    metadata_embeddings.astype("float32")
)

print("FAISS index created!")
print("Vectors stored:", metadata_index.ntotal)
```

```python
def filtered_retrieve(query, filters=None, top_k=5):

    query_embedding = embedder.encode(
        [query],
        convert_to_numpy=True
    ).astype("float32")

    # Search more candidates because some may be removed by filters
    search_k = min(top_k * 3, metadata_index.ntotal)

    distances, indices = metadata_index.search(
        query_embedding,
        search_k
    )

    results = []

    for idx, distance in zip(indices[0], distances[0]):

        if idx == -1:
            continue

        doc = langchain_documents[idx]

        metadata = doc.metadata

        # Apply metadata filters
        if filters:

            # Category filter
            if "category" in filters:
                if metadata.get("category") != filters["category"]:
                    continue

            # Document type filter
            if "document_type" in filters:
                if metadata.get("document_type") != filters["document_type"]:
                    continue

            # Source filter
            if "source" in filters:
                if metadata.get("source") != filters["source"]:
                    continue

            # Date filter
            if "date_after" in filters:
                if metadata.get("date", "") < filters["date_after"]:
                    continue

        results.append({
            "content": doc.page_content,
            "metadata": metadata,
            "distance": float(distance)
        })

        if len(results) >= top_k:
            break

    return results
```

```python
results_no_filter = filtered_retrieve(
    "What database is used by the application?"
)

print("WITHOUT FILTER")
print("=" * 50)

for result in results_no_filter:
    print("\nContent:", result["content"])
    print("Metadata:", result["metadata"])
    print("Distance:", result["distance"])
```

```python
comparison_queries = [
    {
        "query": "Which database is used by the application?",
        "filters": {"category": "technical"},
        "expected_source": "database_guide.pdf"
    },
    {
        "query": "How is user authentication implemented?",
        "filters": {"category": "security"},
        "expected_source": "security_guide.pdf"
    },
    {
        "query": "What is the maximum file size for uploads?",
        "filters": {"category": "product"},
        "expected_source": "user_guide.pdf"
    },
    {
        "query": "What is the refund policy?",
        "filters": {"category": "policy"},
        "expected_source": "refund_policy.pdf"
    },
    {
        "query": "How is the application deployed?",
        "filters": {
            "category": "technical",
            "date_after": "2024-01-01"
        },
        "expected_source": "deployment_guide.pdf"
    }
]

print("Queries:", len(comparison_queries))
```

```python
comparison_results = []

for item in comparison_queries:

    results = filtered_retrieve(
        item["query"],
        filters=None,
        top_k=5
    )

    comparison_results.append({
        "query": item["query"],
        "filters": item["filters"],
        "expected_source": item["expected_source"],
        "without_filters": results
    })

print("Completed unfiltered retrieval.")
```

```python
for i, item in enumerate(comparison_queries):

    results = filtered_retrieve(
        item["query"],
        filters=item["filters"],
        top_k=5
    )

    comparison_results[i]["with_filters"] = results

print("Completed filtered retrieval.")
```

```python
for result in comparison_results:

    print("\n" + "=" * 80)
    print("QUERY:", result["query"])

    print("\nWITHOUT FILTERS:")

    for r in result["without_filters"]:
        print(
            f"- {r['metadata']['source']} "
            f"| {r['metadata']['category']} "
            f"| {r['metadata']['date']} "
            f"| distance={r['distance']:.4f}"
        )

    print("\nWITH FILTERS:")

    for r in result["with_filters"]:
        print(
            f"- {r['metadata']['source']} "
            f"| {r['metadata']['category']} "
            f"| {r['metadata']['date']} "
            f"| distance={r['distance']:.4f}"
        )
```

```python
def calculate_precision(results, expected_source):

    if len(results) == 0:
        return 0.0

    relevant = sum(
        1
        for result in results
        if result["metadata"].get("source") == expected_source
    )

    return relevant / len(results)
```

```python
precision_results = []

for item in comparison_results:

    expected_source = item["expected_source"]

    precision_without = calculate_precision(
        item["without_filters"],
        expected_source
    )

    precision_with = calculate_precision(
        item["with_filters"],
        expected_source
    )

    precision_results.append({
        "query": item["query"],
        "expected_source": expected_source,
        "precision_without_filters": precision_without,
        "precision_with_filters": precision_with
    })

print("Precision calculations completed!")
```

```python
improvement = (
    avg_precision_with -
    avg_precision_without
)

print(
    f"Precision improvement: "
    f"{improvement:+.2f}"
)
```
