```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "google/flan-t5-base"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

print("Model loaded successfully!")
```

```python
def generate_no_context(question):
    prompt = f"Answer this factual question briefly and directly:\n{question}"

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True
    )

    outputs = model.generate(
        **inputs,
        max_new_tokens=100,
        do_sample=False
    )

    response = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return response.strip()
```

```python
documents = [
    "Gold has the chemical symbol Au.",
    "Mars is known as the Red Planet.",
    "The speed of light in vacuum is approximately 299,792 kilometers per second.",
    "Plants primarily absorb carbon dioxide (CO2) from the atmosphere during photosynthesis.",
    "A normal human somatic cell contains 46 chromosomes.",

    "Canberra is the capital city of Australia.",
    "The Pacific Ocean is the largest ocean on Earth.",
    "The Amazon River is the longest river in South America.",
    "Kyoto is a city in Japan.",
    "Ottawa is the capital of Canada.",

    "CPU stands for Central Processing Unit.",
    "A stack is a data structure that follows the LIFO principle.",
    "HTTP stands for Hypertext Transfer Protocol.",
    "Binary search on a sorted array has a time complexity of O(log n).",
    "Python was created by Guido van Rossum.",

    "India gained independence from British rule in 1947.",
    "George Washington was the first President of the United States.",
    "The pyramids at Giza were built by ancient Egyptians.",
    "Jana Gana Mana was written by Rabindranath Tagore.",
    "World War II ended in 1945."
]

print("Number of documents:", len(documents))
```

```python
vectorizer = TfidfVectorizer()

document_vectors = vectorizer.fit_transform(documents)

print("Vector database created successfully.")
```

```python
def retrieve_context(query, top_k=2):
    query_vector = vectorizer.transform([query])

    similarities = cosine_similarity(
        query_vector,
        document_vectors
    )[0]

    top_indices = similarities.argsort()[-top_k:][::-1]

    retrieved = []

    for index in top_indices:
        retrieved.append({
            "document": documents[index],
            "score": float(similarities[index])
        })

    return retrieved
```

```python
query = questions[0]["question"]

retrieved = retrieve_context(query)

print("QUESTION:")
print(query)

print("\nRETRIEVED CONTEXT:")

for item in retrieved:
    print(f"\nScore: {item['score']:.4f}")
    print(item["document"])
```

```python
def generate_rag_answer(question, retrieved_context):

    context = "\n".join(
        item["document"]
        for item in retrieved_context
    )

    prompt = f"""
Answer the question using only the provided context.

Context:
{context}

Question:
{question}

Answer:
"""

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True
    )

    outputs = model.generate(
        **inputs,
        max_new_tokens=100,
        do_sample=False
    )

    response = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return response.strip()
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")
```

```python
import re

def unsupported_claim_check(response, context):

    # Extract numbers
    numbers = re.findall(
        r'\b\d+(?:[.,]\d+)?(?:\s*(?:km|kg|%|years?|million|billion))?\b',
        response,
        flags=re.IGNORECASE
    )

    # Extract dates / years
    dates = re.findall(
        r'\b(?:1[0-9]{3}|20[0-9]{2})\b',
        response
    )

    unsupported_numbers = []

    context_lower = context.lower()

    for item in numbers + dates:
        if item.lower() not in context_lower:
            unsupported_numbers.append(item)

    # Named entities
    doc = nlp(response)

    entities = [
        ent.text
        for ent in doc.ents
        if ent.label_ in [
            "PERSON",
            "ORG",
            "GPE",
            "LOC",
            "EVENT",
            "FAC"
        ]
    ]

    unsupported_entities = []

    for entity in entities:
        if entity.lower() not in context_lower:
            unsupported_entities.append(entity)

    flagged = bool(
        unsupported_numbers or
        unsupported_entities
    )

    return {
        "flagged": flagged,
        "unsupported_numbers": unsupported_numbers,
        "unsupported_entities": unsupported_entities
    }
```

```python
def tokenize(text):
    return set(
        re.findall(
            r'\b[a-zA-Z0-9]+\b',
            text.lower()
        )
    )
```

```python
def jaccard_similarity(response, context):

    response_tokens = tokenize(response)
    context_tokens = tokenize(context)

    if not response_tokens or not context_tokens:
        return 0.0

    intersection = response_tokens & context_tokens
    union = response_tokens | context_tokens

    return len(intersection) / len(union)
```

```python
def retrieval_overlap_check(response, context, threshold=0.15):

    score = jaccard_similarity(
        response,
        context
    )

    return {
        "jaccard_score": score,
        "threshold": threshold,
        "flagged": score < threshold
    }
```

```python
def contradiction_check(response, context):

    prompt = f"""
Determine whether the answer contradicts the provided context.

Context:
{context}

Answer:
{response}

Does the answer contradict the context?
Answer only YES or NO.
"""

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True
    )

    outputs = model.generate(
        **inputs,
        max_new_tokens=5,
        do_sample=False
    )

    result = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    ).strip().lower()

    flagged = result.startswith("yes")

    return {
        "model_judgement": result,
        "flagged": flagged
    }
```

```python
def citation_absence_check(response):

    citation_patterns = [
        r'\[[0-9]+\]',
        r'\(source',
        r'\(reference',
        r'\baccording to\b',
        r'\bsource:',
        r'\breference:'
    ]

    has_citation = any(
        re.search(
            pattern,
            response,
            flags=re.IGNORECASE
        )
        for pattern in citation_patterns
    )

    # Any non-empty factual answer without citation
    factual_assertion = len(
        tokenize(response)
    ) > 2

    flagged = factual_assertion and not has_citation

    return {
        "has_citation": has_citation,
        "flagged": flagged
    }
```

```python
def calculate_confidence(signals):

    total_signals = 4

    flagged_signals = sum(
        signal["flagged"]
        for signal in signals.values()
    )

    confidence = (
        1 - flagged_signals / total_signals
    ) * 100

    return confidence
```

```python
def detect_hallucination(result):

    response = result["rag_response"]

    context = "\n".join(
        result["retrieved_context"]
    )

    signal_1 = unsupported_claim_check(
        response,
        context
    )

    signal_2 = retrieval_overlap_check(
        response,
        context
    )

    signal_3 = contradiction_check(
        response,
        context
    )

    signal_4 = citation_absence_check(
        response
    )

    signals = {
        "unsupported_claim": signal_1,
        "retrieval_overlap": signal_2,
        "contradiction": signal_3,
        "citation_absence": signal_4
    }

    confidence = calculate_confidence(
        signals
    )

    return {
        "signals": signals,
        "final_confidence_score": confidence
    }
```
