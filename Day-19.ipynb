```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct",
    device=-1
)

print("Model loaded successfully!")
```

```python
test_data = [
    {
        "context": "Python was created by Guido van Rossum and was first released in 1991.",
        "question": "Who created Python?",
        "answer": "Guido van Rossum"
    },
    {
        "context": "The Amazon River is approximately 6,400 kilometers long and flows through South America.",
        "question": "Where does the Amazon River flow?",
        "answer": "South America"
    },
    {
        "context": "Machine learning is a subset of artificial intelligence that allows systems to learn patterns from data.",
        "question": "What is machine learning?",
        "answer": "A subset of artificial intelligence that learns patterns from data"
    },
    {
        "context": "HTTP is a protocol used for transferring resources over the web.",
        "question": "What is HTTP used for?",
        "answer": "Transferring resources over the web"
    },
    {
        "context": "SQL is a language commonly used to manage and query relational databases.",
        "question": "What is SQL used for?",
        "answer": "Managing and querying relational databases"
    },
    {
        "context": "The human heart has four chambers: two atria and two ventricles.",
        "question": "How many chambers does the human heart have?",
        "answer": "Four"
    },
    {
        "context": "Docker packages applications and their dependencies into containers.",
        "question": "What does Docker package?",
        "answer": "Applications and their dependencies"
    },
    {
        "context": "Cloud computing provides computing resources such as storage and processing power over the internet.",
        "question": "What does cloud computing provide?",
        "answer": "Computing resources over the internet"
    },
    {
        "context": "Git is a distributed version control system used to track changes in source code.",
        "question": "What is Git used for?",
        "answer": "Tracking changes in source code"
    },
    {
        "context": "FAISS is a library designed for efficient similarity search of dense vectors.",
        "question": "What is FAISS designed for?",
        "answer": "Efficient similarity search of dense vectors"
    }
]

print("Number of test cases:", len(test_data))
```

```python
baseline_prompt = """
Answer the question using the context provided.

Context:
{context}

Question:
{question}

Answer:
"""

print(baseline_prompt)
```

```python
def ask_model(prompt):
    result = generator(
        prompt,
        max_new_tokens=100,
        do_sample=False
    )

    return result[0]["generated_text"].strip()

print("Model function created successfully!")
```

```python
def ask_model(prompt):
    result = generator(
        prompt,
        max_new_tokens=100,
        do_sample=False
    )

    generated = result[0]["generated_text"]

    # Remove the original prompt from the returned text
    if generated.startswith(prompt):
        generated = generated[len(prompt):]

    return generated.strip()

print("Model function created successfully!")
```

```python
test_prompt = baseline_prompt.format(
    context=test_data[0]["context"],
    question=test_data[0]["question"]
)

print(ask_model(test_prompt))
```

```python
import pandas as pd
```

```python
baseline_results = []

for item in test_data:
    prompt = baseline_prompt.format(
        context=item["context"],
        question=item["question"]
    )

    output = ask_model(prompt)

    baseline_results.append({
        "question": item["question"],
        "expected": item["answer"],
        "output": output
    })

baseline_df = pd.DataFrame(baseline_results)

baseline_df
```

```python
def calculate_accuracy(expected, output):
    expected_words = set(expected.lower().split())
    output_words = set(output.lower().split())

    if expected.lower() in output.lower():
        return 5

    overlap = len(expected_words & output_words) / max(len(expected_words), 1)

    if overlap >= 0.75:
        return 4
    elif overlap >= 0.5:
        return 3
    elif overlap > 0:
        return 2
    else:
        return 1
```

```python
def calculate_format_score(output):
    word_count = len(output.split())

    if word_count <= 10:
        return 5
    elif word_count <= 20:
        return 4
    elif word_count <= 35:
        return 3
    elif word_count <= 50:
        return 2
    else:
        return 1
```

```python
baseline_df["accuracy"] = baseline_df.apply(
    lambda row: calculate_accuracy(
        row["expected"],
        row["output"]
    ),
    axis=1
)

baseline_df["format_consistency"] = baseline_df["output"].apply(
    calculate_format_score
)

baseline_df
```

```python
baseline_accuracy = baseline_df["accuracy"].mean()
baseline_format = baseline_df["format_consistency"].mean()

baseline_overall = (
    baseline_accuracy + baseline_format
) / 2

print("Baseline Accuracy:", round(baseline_accuracy, 2))
print("Baseline Format Consistency:", round(baseline_format, 2))
print("Baseline Overall Score:", round(baseline_overall, 2))
```

```python
# Technique 1: Role Assignment
role_prompt = """
You are an expert question-answering assistant.

Answer the question using only the provided context.
Be accurate and concise.

Context:
{context}

Question:
{question}

Answer:
"""


# Technique 2: Output Format Specification
format_prompt = """
Answer the question using only the provided context.

Return the answer in exactly this format:

Answer: <short answer>

Context:
{context}

Question:
{question}
"""


# Technique 3: Reasoning
reasoning_prompt = """
Answer the question using only the provided context.

Before answering, carefully verify that the answer is supported
by the context. Do not provide your reasoning process.
Return only the final concise answer.

Context:
{context}

Question:
{question}

Final answer:
"""


# Technique 4: Few-Shot Examples
fewshot_prompt = """
Answer questions using only the provided context.

Example 1:

Context:
Python was created by Guido van Rossum.

Question:
Who created Python?

Answer:
Guido van Rossum

Example 2:

Context:
Git is a version control system used to track source-code changes.

Question:
What is Git used for?

Answer:
Tracking source-code changes

Now answer the following question.

Context:
{context}

Question:
{question}

Answer:
"""


# Technique 5: Negative Constraints
negative_prompt = """
Answer the question using only the provided context.

Rules:
- Do not use information outside the context.
- Do not make assumptions.
- Do not invent facts.
- Do not provide unrelated information.
- Keep the answer concise.
- If the answer is not supported by the context, say:
  "Not supported by the provided context."

Context:
{context}

Question:
{question}

Answer:
"""

print("All five prompt techniques created successfully!")
```

```python
prompts = {
    "baseline": baseline_prompt,
    "role_assignment": role_prompt,
    "output_format": format_prompt,
    "reasoning": reasoning_prompt,
    "few_shot": fewshot_prompt,
    "negative_constraints": negative_prompt
}

print("Number of prompt versions:", len(prompts))
```

```python
for name in prompts:
    print("-", name)
```

```python
def evaluate_prompt(prompt_template, test_data):
    results = []

    for item in test_data:
        prompt = prompt_template.format(
            context=item["context"],
            question=item["question"]
        )

        output = ask_model(prompt)

        results.append({
            "question": item["question"],
            "expected": item["answer"],
            "output": output
        })

    return pd.DataFrame(results)

print("Evaluation function created successfully!")
```

```python
role_results["accuracy"] = role_results.apply(
    lambda row: calculate_accuracy(
        row["expected"],
        row["output"]
    ),
    axis=1
)

role_results["format_consistency"] = role_results["output"].apply(
    calculate_format_score
)

role_results
```

```python
role_accuracy = role_results["accuracy"].mean()
role_format = role_results["format_consistency"].mean()
role_overall = (role_accuracy + role_format) / 2

print("Role Assignment Accuracy:", round(role_accuracy, 2))
print("Role Assignment Format:", round(role_format, 2))
print("Role Assignment Overall:", round(role_overall, 2))
```

```python
grounding_system_prompt = """
You are a grounded RAG assistant.

Your answers must be based ONLY on the information provided in the context.

Rules:
1. Use only the provided context to answer the question.
2. Do not use outside knowledge or information from your training data.
3. Do not guess or make assumptions.
4. If the answer is not directly supported by the context, refuse to answer.
5. For unsupported questions, respond exactly:
   "I cannot answer this question because the provided context does not contain enough information."
6. Keep supported answers concise.
"""

print(grounding_system_prompt)
```

```python
out_of_context_questions = [
    "Who is the current President of the United States?",
    "What is the capital of Australia?",
    "Who founded Microsoft?",
    "What is the population of India?",
    "Who won the FIFA World Cup in 2022?"
]

print("Number of out-of-context questions:", len(out_of_context_questions))
```

```python
context = "\n".join(
    [item["context"] for item in test_data]
)

print(context)
```

```python
def grounded_answer(context, question):

    prompt = f"""
{grounding_system_prompt}

Context:
{context}

Question:
{question}

Answer:
"""

    return ask_model(prompt)

print("Grounding function created successfully!")
```

```python
grounding_results = []

for question in out_of_context_questions:

    answer = grounded_answer(
        context,
        question
    )

    grounding_results.append({
        "Question": question,
        "Model Response": answer
    })

grounding_df = pd.DataFrame(grounding_results)

grounding_df
```

```python
def check_refusal(response):
    response = response.lower()

    refusal_phrases = [
        "cannot answer",
        "not enough information",
        "not supported",
        "not provided",
        "cannot be determined",
        "insufficient information"
    ]

    return any(
        phrase in response
        for phrase in refusal_phrases
    )
```

```python
grounding_df["Refused Correctly"] = grounding_df[
    "Model Response"
].apply(check_refusal)

grounding_df
```

```python
grounding_accuracy = (
    grounding_df["Refused Correctly"].mean() * 100
)

print(
    f"Grounding refusal accuracy: {grounding_accuracy:.1f}%"
)
```
