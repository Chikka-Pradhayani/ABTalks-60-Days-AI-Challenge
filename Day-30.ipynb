```python
import re

def normalize(text):
    text = text.lower()
    text = re.sub(r"[^a-z0-9\s]", "", text)
    return set(text.split())

def score(answer, expected):
    answer_words = normalize(answer)
    expected_words = normalize(expected)

    if not answer_words or not expected_words:
        return 1

    ratio = len(answer_words & expected_words) / len(expected_words)

    if ratio >= 0.9:
        return 5
    if ratio >= 0.75:
        return 4
    if ratio >= 0.55:
        return 3
    if ratio >= 0.3:
        return 2
    return 1

def evaluate_report(question, context, answer, ground_truth):
    return {
        "groundedness": score(answer, context),
        "correctness": score(answer, ground_truth),
        "completeness": score(answer, ground_truth)
    }
```

```python
page = '''
"use client";

import { useState } from "react";

const steps = [
  "Researching the topic",
  "Analyzing gathered information",
  "Generating the report",
  "Evaluating the report"
];

export default function Home() {
  const [topic, setTopic] = useState("");
  const [loading, setLoading] = useState(false);
  const [currentStep, setCurrentStep] = useState(0);
  const [data, setData] = useState(null);

  async function research() {
    if (!topic.trim()) return;

    setLoading(true);
    setData(null);
    setCurrentStep(1);

    const timer = setInterval(() => {
      setCurrentStep((step) => {
        if (step < 4) return step + 1;
        return step;
      });
    }, 1200);

    try {
      const response = await fetch(
        `${process.env.NEXT_PUBLIC_API_URL}/research`,
        {
          method: "POST",
          headers: {
            "Content-Type": "application/json"
          },
          body: JSON.stringify({ topic })
        }
      );

      const result = await response.json();

      clearInterval(timer);
      setCurrentStep(4);
      setData(result);
    } catch (error) {
      clearInterval(timer);
      console.error(error);
    }

    setLoading(false);
  }

  return (
    <main style={{
      maxWidth: "900px",
      margin: "40px auto",
      padding: "20px",
      fontFamily: "Arial"
    }}>
      <h1>AI Research Assistant</h1>

      <div>
        <input
          value={topic}
          onChange={(e) => setTopic(e.target.value)}
          placeholder="Enter a research topic"
          style={{
            width: "70%",
            padding: "12px",
            marginRight: "10px"
          }}
        />

        <button
          onClick={research}
          disabled={loading}
          style={{ padding: "12px 20px" }}
        >
          {loading ? "Researching..." : "Research"}
        </button>
      </div>

      {(loading || data) && (
        <div style={{ marginTop: "30px" }}>
          <h2>Research Workflow</h2>

          {steps.map((step, index) => {
            const stepNumber = index + 1;
            const completed = currentStep >= stepNumber;

            return (
              <div
                key={step}
                style={{
                  padding: "12px",
                  margin: "8px 0",
                  border: "1px solid #ddd",
                  borderRadius: "8px"
                }}
              >
                {completed ? "✓" : "○"} {step}
              </div>
            );
          })}
        </div>
      )}

      {data && (
        <div style={{ marginTop: "30px" }}>
          <h2>Research Report</h2>

          <pre style={{
            whiteSpace: "pre-wrap",
            lineHeight: "1.6"
          }}>
            {data.report}
          </pre>

          <h2>Evaluation</h2>

          <p>
            Groundedness: {data.evaluation.groundedness}/5
          </p>

          <p>
            Correctness: {data.evaluation.correctness}/5
          </p>

          <p>
            Completeness: {data.evaluation.completeness}/5
          </p>
        </div>
      )}
    </main>
  );
}
'''

with open(
    "day30_ai_research_assistant/frontend/app/page.js",
    "w"
) as f:
    f.write(page)

print("Real-time workflow progress added")
```

```python
import requests
import json

evaluation_dataset = [
    {
        "id": 1,
        "question": "What is artificial intelligence?",
        "ground_truth": "Artificial intelligence is the ability of machines to perform tasks that normally require human intelligence."
    },
    {
        "id": 2,
        "question": "What is machine learning?",
        "ground_truth": "Machine learning is a branch of artificial intelligence where systems learn patterns from data."
    },
    {
        "id": 3,
        "question": "What is deep learning?",
        "ground_truth": "Deep learning is a machine learning approach that uses neural networks with multiple layers."
    },
    {
        "id": 4,
        "question": "What is natural language processing?",
        "ground_truth": "Natural language processing enables computers to understand and process human language."
    },
    {
        "id": 5,
        "question": "What are embeddings?",
        "ground_truth": "Embeddings are numerical vector representations of text or other data."
    },
    {
        "id": 6,
        "question": "What is RAG?",
        "ground_truth": "Retrieval augmented generation combines information retrieval with language generation."
    },
    {
        "id": 7,
        "question": "What is a hallucination in AI?",
        "ground_truth": "An AI hallucination is an incorrect or unsupported response generated by an AI system."
    },
    {
        "id": 8,
        "question": "What is prompt engineering?",
        "ground_truth": "Prompt engineering is the process of designing instructions that guide an AI model to produce useful responses."
    },
    {
        "id": 9,
        "question": "What is a vector database?",
        "ground_truth": "A vector database stores and searches numerical vector representations efficiently."
    },
    {
        "id": 10,
        "question": "What is FAISS?",
        "ground_truth": "FAISS is a library for efficient similarity search and clustering of dense vectors."
    },
    {
        "id": 11,
        "question": "What is cosine similarity?",
        "ground_truth": "Cosine similarity measures the similarity between vectors based on the angle between them."
    },
    {
        "id": 12,
        "question": "What is text chunking?",
        "ground_truth": "Text chunking divides large documents into smaller pieces for processing and retrieval."
    },
    {
        "id": 13,
        "question": "What is FastAPI?",
        "ground_truth": "FastAPI is a Python web framework for building APIs."
    },
    {
        "id": 14,
        "question": "What is an API?",
        "ground_truth": "An API is an interface that allows software applications to communicate with each other."
    },
    {
        "id": 15,
        "question": "What is a language model?",
        "ground_truth": "A language model is an AI model that learns patterns in language to understand or generate text."
    },
    {
        "id": 16,
        "question": "What is data preprocessing?",
        "ground_truth": "Data preprocessing prepares raw data by cleaning and transforming it before analysis or machine learning."
    },
    {
        "id": 17,
        "question": "What is tokenization?",
        "ground_truth": "Tokenization divides text into smaller units called tokens."
    },
    {
        "id": 18,
        "question": "What is TF-IDF?",
        "ground_truth": "TF-IDF measures the importance of a word in a document relative to a collection of documents."
    },
    {
        "id": 19,
        "question": "What is AI evaluation?",
        "ground_truth": "AI evaluation measures the quality and performance of an AI system using defined criteria."
    },
    {
        "id": 20,
        "question": "Why is AI evaluation important?",
        "ground_truth": "AI evaluation helps identify weaknesses and measure whether an AI system produces reliable results."
    }
]

def normalize(text):
    import re
    text = text.lower()
    text = re.sub(r"[^a-z0-9\s]", "", text)
    return set(text.split())

def score(answer, expected):
    answer_words = normalize(answer)
    expected_words = normalize(expected)

    if not answer_words or not expected_words:
        return 1

    ratio = len(answer_words & expected_words) / len(expected_words)

    if ratio >= 0.9:
        return 5
    elif ratio >= 0.75:
        return 4
    elif ratio >= 0.55:
        return 3
    elif ratio >= 0.3:
        return 2

    return 1

results = []

for item in evaluation_dataset:
    response = requests.post(
        "http://127.0.0.1:8000/research",
        json={"topic": item["question"]}
    )

    data = response.json()
    answer = data["report"]

    scores = {
        "groundedness": score(answer, item["ground_truth"]),
        "correctness": score(answer, item["ground_truth"]),
        "completeness": score(answer, item["ground_truth"])
    }

    results.append({
        "id": item["id"],
        "question": item["question"],
        "scores": scores
    })

print("Evaluation completed")
print("Questions evaluated:", len(results))
```

```python
averages = {
    "groundedness": sum(
        r["scores"]["groundedness"] for r in results
    ) / len(results),

    "correctness": sum(
        r["scores"]["correctness"] for r in results
    ) / len(results),

    "completeness": sum(
        r["scores"]["completeness"] for r in results
    ) / len(results)
}

baseline = {
    key: round(value, 2)
    for key, value in averages.items()
}

print("Official baseline:")
print(json.dumps(baseline, indent=2))
```

```python
import os
import subprocess

project = "day30_ai_research_assistant"
backend = os.path.join(project, "backend")

requirements = """fastapi
uvicorn
pydantic
"""

with open(os.path.join(backend, "requirements.txt"), "w") as f:
    f.write(requirements)

print("requirements.txt created")

subprocess.run(
    ["git", "add", "."],
    cwd=project
)

subprocess.run(
    ["git", "commit", "-m", "Prepare backend for deployment"],
    cwd=project
)

result = subprocess.run(
    ["git", "push"],
    cwd=project,
    capture_output=True,
    text=True
)

print(result.stdout)

if result.returncode != 0:
    print(result.stderr)
else:
    print("Deployment files pushed to GitHub")
```

```python
import os
import subprocess

project = "day30_ai_research_assistant"
frontend = os.path.join(project, "frontend")

backend_url = "https://YOUR-RENDER-SERVICE.onrender.com"

with open(os.path.join(frontend, ".env.local"), "w") as f:
    f.write(f"NEXT_PUBLIC_API_URL={backend_url}\n")

subprocess.run(
    ["git", "add", "."],
    cwd=project
)

subprocess.run(
    ["git", "commit", "-m", "Connect frontend to deployed backend"],
    cwd=project
)

result = subprocess.run(
    ["git", "push"],
    cwd=project,
    capture_output=True,
    text=True
)

print(result.stdout)

if result.returncode != 0:
    print(result.stderr)
else:
    print("Frontend API configuration committed")
```
